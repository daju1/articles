**Задача.** В §4 постановки тензор записан как $\langle T_{ik}\rangle=\frac{\hbar}{\pi}\int_0^\infty d\xi\,\mathcal T_{ik}(\xi)$ без нормировочных множителей. Здесь множители выводятся от начала до конца и проверяются тремя независимыми способами на задаче, ответ которой известен точно: давление между двумя идеально проводящими пластинами на расстоянии $a$,
$$
P=-\frac{\pi^2\hbar c}{240\,a^4}\qquad\text{(Казимир 1948).}
$$

Три проверки:
1. **аналитически** — функция Грина пластин методом изображений, сумма по изображениям — точно;
2. **формула Лифшица** — независимая формула через коэффициенты отражения $r=\pm1$;
3. **численно** — тем же решателем на панелях, которым считался уголок (`../casimir/corner_lib.py`), пластины конечной длины $L$.

Разделы: 1 — квантование и вакуумные средние; 2 — поворот Вика и нормировка; 3 — сведение к двум скалярам для $z$-однородной геометрии; 4–6 — три проверки; 7 — итог для постановки.

## 1. Вакуумные средние поля

Гауссова система. Каждая мода $\nu$ поля в основном состоянии несёт энергию $\hbar\omega_\nu/2$ — это нулевые колебания Казимира. Вакуумные средние в совпадающих точках выражаются через функцию Грина на мнимой частоте $\omega=i\xi$; ключевое тождество (см. также `Casimir_cylinders_Maxwell_general.ipynb` и ответ о связи с суммой Казимира):
$$
\frac{1}{\pi}\int_0^\infty\frac{d\xi}{\omega_\nu^2+\xi^2}=\frac{1}{2\omega_\nu},
$$
так что интеграл по $\xi$ от разложения функции Грина по модам $\sum_\nu\frac{u_\nu(\mathbf r)u^*_\nu(\mathbf r')}{\omega_\nu^2+\xi^2}$ воспроизводит почленно $\sum_\nu\frac{\hbar}{2\omega_\nu}u_\nu u_\nu^*$ — вклад нулевых колебаний.

Для электромагнитного поля общая формула — через тензорную функцию Грина $\hat{\mathcal G}$ уравнения $[\mathrm{rot}\,\mathrm{rot}+\xi^2/c^2]\hat{\mathcal G}=\hat1\delta$ (Лифшиц–Питаевский, т. IX, §§ о силах Ван-дер-Ваальса):
$$
\big\langle E_iE_k\big\rangle_s\propto\frac{\hbar}{\pi}\int_0^\infty d\xi\,\frac{\xi^2}{c^2}\,\mathcal G_{s,ik}(\mathbf r,\mathbf r;\xi),\qquad
\big\langle H_iH_k\big\rangle_s\propto\frac{\hbar}{\pi}\int_0^\infty d\xi\,\big[\mathrm{rot}\,\hat{\mathcal G}_s\,\overleftarrow{\mathrm{rot}}'\big]_{ik}.
$$
**Множители и знаки в этой электромагнитной форме здесь не выводятся и не проверяются** — она понадобится для неидеальных металлов (проверка П3). Для идеального $z$-однородного проводника задача сводится к двум скалярам (§2), и именно скалярная формула проверяется ниже тремя способами.


## 2. Тензор и его вид для $z$-однородной геометрии

$$
T_{ik}=\frac1{4\pi}\Big[E_iE_k+H_iH_k-\tfrac12\delta_{ik}\big(E^2+H^2\big)\Big],\qquad f_i=\oint_C\langle T_{ik}\rangle n_k\,dl .
$$

**Сведение к двум скалярам** (для идеального проводника, тело однородно по $z$). Поле раскладывается на ТМ ($E_z=\varphi$, Дирихле) и ТЕ ($H_z=\psi$, Нейман). Каждая поляризация — безмассовое скалярное поле в 3D со своим условием; на мнимой частоте при фиксированных $(\xi,k_z)$ оператор зависит только от $\kappa^2=\xi^2/c^2+k_z^2$:
$$
(-\nabla_\perp^2+\kappa^2)\,g(\mathbf r,\mathbf r';\kappa)=\delta^2(\mathbf r-\mathbf r'),\qquad g|_{\partial D}=0\ \ (\text{ТМ}),\quad \partial_ng|_{\partial D}=0\ \ (\text{ТЕ}).
$$
Тензор каждой поляризации — тензор скалярного поля $\Theta_{ij}=\partial_i\varphi\partial_j\varphi-\tfrac12\delta_{ij}\big((\nabla\varphi)^2-\dot\varphi^2/c^2\big)$, а Максвелловский $\langle T_{ij}\rangle=-\big(\langle\Theta^{\rm TM}_{ij}\rangle+\langle\Theta^{\rm TE}_{ij}\rangle\big)$ в смысле полной силы (знак: $T$ — натяжение, $\Theta$ — поток импульса). Для скаляра
$$
\boxed{\ \big\langle\Theta_{ij}(\mathbf r)\big\rangle=\hbar c\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}\Big[D_{ij}-\tfrac12\delta_{ij}\big(D_{kk}+\kappa^2g_s\big)\Big],\qquad D_{ij}=\partial_i\partial'_jg_s\big|_{\mathbf r'=\mathbf r}\ }
$$
где мера $\int\frac{d\xi\,dk_z}{(2\pi)^2}\to\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}$ (полярные координаты в плоскости $(\xi/c,k_z)$, $\xi>0$ и $\xi<0$ вместе), а $\hbar c$ — из $\langle\varphi\varphi\rangle=\hbar c\int\frac{d\xi\,dk_z}{(2\pi)^2}\,g$ для поля с лагранжианом $\frac{1}{2c^2}\dot\varphi^2-\frac12(\nabla\varphi)^2$ (это проверяется в §3–§5). Это и есть формула, которой считались все предыдущие ноутбуки ($\hbar=c=1$).

**Что проверяем.** Сам множитель $\hbar c\cdot\frac{\kappa d\kappa}{2\pi}$ и знаки. Если они верны, для двух пластин каждая поляризация даёт $\langle\Theta_{yy}\rangle=-\pi^2\hbar c/480a^4$, а сумма $-\pi^2\hbar c/240a^4$.

**Почему ТЕ+ТМ = ЭМ для пластин точно.** Для двух параллельных пластин моды ЭМ поля — это моды Дирихле $\sin(n\pi y/a)$ с $n\ge1$ (ТМ) и Неймана $\cos(n\pi y/a)$ с $n\ge0$ (ТЕ). Мода $n=0$ Неймана не зависит от $a$ и в силу не входит. Поэтому сумма двух скаляров — это в точности Казимир.

## 3. Пластины методом изображений

Пластины $y=0$ и $y=a$, бесконечные по $x$ и $z$. Делаем фурье по $x$ ($k_x$): одномерная задача для $q^2=\kappa^2+k_x^2$, $(-\partial_y^2+q^2)g=\delta(y-y')$. Свободное решение $e^{-q|y-y'|}/2q$, изображения:
$$
g(y,y')=\frac{1}{2q}\sum_{n=-\infty}^{\infty}\Big[e^{-q|y-y'+2na|}\mp e^{-q|y+y'+2na|}\Big]\qquad(-\ \text{Дирихле},\ +\ \text{Нейман}).
$$
Рассеянная часть — всё, кроме $n=0$ в первой сумме. В совпадающих точках:
$$
g_s=\frac1{2q}\Big[\sum_{n\ne0}e^{-2q|n|a}\mp\sum_ne^{-q|2y+2na|}\Big],\qquad
\partial_y\partial_{y'}g_s=-\frac q2\sum_{n\ne0}e^{-2q|n|a}\pm\frac q2\sum_ne^{-q|2y+2na|},\qquad
\partial_x\partial_{x'}g_s=k_x^2g_s .
$$
Меру $\int\frac{\kappa d\kappa}{2\pi}\int\frac{dk_x}{2\pi}$ переписываем как $\int\frac{d^3q}{(2\pi)^3}$ (вектор $(\xi/c,k_z,k_x)$), $k_x^2\to q^2/3$ по углам.

**Что видно до расчёта.** Члены с $e^{-q|2y+2na|}$ («отражённые изображения») зависят от $y$ и расходятся у стенок. В $\Theta_{yy}$ они **сокращаются тождественно**: $D_{yy}-\tfrac12(D_{yy}+D_{xx}+\kappa^2g_s)=\tfrac12\big(D_{yy}-q^2g_s\big)$, а у отражённой части $D_{yy}=+q^2g_s$. Поэтому $\Theta_{yy}$ не зависит от $y$ — как и должно быть по $\partial_y\Theta_{yy}=0$. В $\Theta_{xx}$ они остаются, расходятся у стенок по-разному для Дирихле и Неймана и сокращаются только в сумме ТМ+ТЕ.

```sage
import numpy as np
pi = np.pi

def theta_plates(y, a, pol, nq=4000, nimg=60):
    """<Theta_yy>, <Theta_xx> скалярного поля между пластинами y=0,a; hbar=c=1. pol='D' или 'N'."""
    s = -1 if pol == 'D' else +1
    n = np.arange(-nimg, nimg+1); nz = n[n != 0]
    u = (np.arange(nq)+0.5)/nq; q = np.tan(pi/2*u)/a; w = pi/2/np.cos(pi/2*u)**2/nq/a
    Tyy = np.zeros(nq); Txx = np.zeros(nq)
    for j, qq in enumerate(q):
        Eper = np.exp(-2*qq*np.abs(nz)*a).sum()
        Eimg = np.exp(-qq*np.abs(2*y + 2*n*a)).sum()
        gs  = Eper/(2*qq) + s*Eimg/(2*qq)
        Dyy = -(qq/2)*Eper + s*(qq/2)*Eimg
        kx2 = qq*qq/3; Dxx = kx2*gs; kap2 = qq*qq - kx2
        Tyy[j] = Dyy - 0.5*(Dxx + Dyy + kap2*gs)
        Txx[j] = Dxx - 0.5*(Dxx + Dyy + kap2*gs)
    m = q**2*w/(2*pi**2)          # d^3q/(2pi)^3 = 4 pi q^2 dq/(2pi)^3
    return (Tyy*m).sum(), (Txx*m).sum()

a = 1.0
print('цели: Theta_yy одной поляризации = -pi^2/480 = %.6f;  сумма = -pi^2/240 = %.6f;  Theta_xx суммы = +pi^2/720 = %.6f'
      % (-pi**2/480, -pi**2/240, pi**2/720))
for y in (0.2, 0.5, 0.8):
    D = theta_plates(y, a, 'D'); N = theta_plates(y, a, 'N')
    print('y=%.1f  Дирихле (yy,xx)=(%+.6f,%+.6f)  Нейман (%+.6f,%+.6f)  сумма (%+.6f,%+.6f)' % (y, *D, *N, D[0]+N[0], D[1]+N[1]))
```

**Результат (у меня):**

| $y/a$ | Дирихле $\Theta_{yy}$ | Нейман $\Theta_{yy}$ | сумма $\Theta_{yy}$ | Дирихле $\Theta_{xx}$ | Нейман $\Theta_{xx}$ | сумма $\Theta_{xx}$ |
|---|---|---|---|---|---|---|
| 0,2 | $-0{,}020562$ | $-0{,}020562$ | $-0{,}041123$ | $+3{,}98436$ | $-3{,}97065$ | $+0{,}013708$ |
| 0,5 | $-0{,}020562$ | $-0{,}020562$ | $-0{,}041123$ | $+0{,}21247$ | $-0{,}19876$ | $+0{,}013708$ |
| 0,8 | $-0{,}020562$ | $-0{,}020562$ | $-0{,}041123$ | $+3{,}98436$ | $-3{,}97065$ | $+0{,}013708$ |
| точно | $-\pi^2/480$ | $-\pi^2/480$ | $-\pi^2/240=-0{,}041123$ | | | $+\pi^2/720=0{,}013708$ |

- $\Theta_{yy}$ совпадает с $-\pi^2/480$ для каждой поляризации и с $-\pi^2/240$ в сумме **до 6 знаков** и не зависит от $y$.
- Сумма ТМ+ТЕ: $\langle\Theta_{ij}\rangle=\frac{\pi^2\hbar c}{720a^4}\,\mathrm{diag}(+1,-3,+1)$ в осях $(x,y,z)$ — это известный тензор Брауна–Маклэя.
- По отдельности $\Theta_{xx}$ расходится у стенок с противоположными знаками для Дирихле и Неймана — поэтому локальное давление на кромки и грани отдельной поляризации (§12b–12e старого ноутбука) не имеет смысла без второй поляризации.

**Вывод:** множитель $\hbar c\int\frac{\kappa d\kappa}{2\pi}$ и знаки в формуле §2 верны.

## 4. Независимая проверка: формула Лифшица

Формула Лифшица (Лифшиц–Питаевский, т. IX) для давления между полупространствами с коэффициентами отражения $r_{\rm TM},r_{\rm TE}$ на мнимой частоте:
$$
P=-\frac{\hbar}{2\pi^2}\int_0^\infty d\xi\int_0^\infty k_\perp dk_\perp\,\kappa\sum_{p}\frac{r_p^2e^{-2\kappa a}}{1-r_p^2e^{-2\kappa a}},\qquad \kappa^2=\xi^2/c^2+k_\perp^2 .
$$
Для идеального проводника $r_{\rm TM}=1$, $r_{\rm TE}=-1$, $r_p^2=1$. Это **другой** путь: нет функции Грина в совпадающих точках, нет изображений, нет нашей нормировки.

```sage
def lifshitz_ideal(a, n=4000):
    """P для идеальных пластин из формулы Лифшица; hbar=c=1. В полярных (xi,k_perp): d xi k dk -> K^2 dK."""
    u = (np.arange(n)+0.5)/n; K = np.tan(pi/2*u)/a; w = pi/2/np.cos(pi/2*u)**2/n/a
    with np.errstate(over='ignore'):
        f = 2*K/np.expm1(2*K*a)          # kappa * sum_p r^2 e/(1-r^2 e), две поляризации
    return -(1/(2*pi**2))*(K**2*w*f).sum()
print('Лифшиц: P = %.10f;  -pi^2/240 = %.10f' % (lifshitz_ideal(1.0), -pi**2/240))
```

**Результат (у меня):** $P=-0{,}0411233517$ при $-\pi^2/240=-0{,}0411233517$ — совпадение до $10^{-10}$.

Две независимые формулы (функция Грина в совпадающих точках и формула Лифшица через $r_p$) дают одно и то же. Нормировка §2 подтверждена.

## 5. Проверка решателя на панелях: пластины конечной длины

Тот же код, что для уголка (`../casimir/corner_lib.py`, функция `tensor_l` — Дирихле, простой слой). Две тонкие пластины длиной $L$ при $y=0$ и $y=a$, $N_l$ панелей на каждой; давление $-\langle\Theta_{yy}\rangle$ в центре зазора сравнивается с $\pi^2\hbar c/480a^4$.

Ожидание: при $L\gg a$ отношение $\to1$, а отклонение уменьшается с размером панели $h_{\max}$.

Внимание: при $N_l\ge480$ нужна память устройства; у меня считалось на Mac, около минуты на точку.

```sage
import os
G = {}
exec(open('../casimir/corner_lib.py').read(), G)          # panels, tensor_l, ...
a = 1.0; ideal = pi**2/480/a**4; pt = np.array([[0.0, a/2]])
for (L, Nl, nk) in ((6.0, 240, 60), (12.0, 480, 60), (24.0, 960, 60), (12.0, 240, 60), (12.0, 960, 60), (12.0, 480, 120)):
    B = [G['panels']([-L/2, 0], [L/2, 0], Nl), G['panels']([-L/2, a], [L/2, a], Nl)]
    p = -G['tensor_l'](B, pt, None, nk=nk, smin=1/a)[0, 1, 1]
    print('L=%4.0f  Nl=%4d  h_max=%.4f  nk=%3d   p/ideal = %.5f' % (L, Nl, B[0][1].max(), nk, p/ideal))
```

**Результат (у меня):**

| $L/a$ | $N_l$ | $h_{\max}/a$ | $n_\kappa$ | $p/p_{\rm ideal}$ |
|---|---|---|---|---|
| 6 | 240 | 0,039 | 60 | 1,0074 |
| 12 | 480 | 0,039 | 60 | 1,0073 |
| 24 | 960 | 0,039 | 60 | 1,0073 |
| 12 | 240 | 0,079 | 60 | 1,0145 |
| 12 | 480 | 0,039 | 60 | 1,0073 |
| 12 | 960 | 0,020 | 60 | 1,0037 |
| 12 | 480 | 0,039 | 120 | 1,0073 |

- При фиксированном шаге $h_{\max}$ результат **не зависит** от $L$ (6, 12, 24): краевые эффекты в центре уже не видны.
- Не зависит от сетки по $\kappa$ (60 и 120 узлов).
- Отклонение **пропорционально шагу**: $1{,}45\%\to0{,}73\%\to0{,}37\%$ при $h_{\max}$ = 0,079 → 0,039 → 0,020. Это первый порядок кусочно-постоянной плотности на панелях, предел $h\to0$ даёт $1{,}0000$.

**Вывод:** решатель на панелях воспроизводит Казимира с ошибкой $\approx0{,}19\,h_{\max}/a$; нормировка в нём та же, что в §2.

## 6. Итог для постановки

1. Нормировка (вставить в §4 `Antipin_corner_problem_statement.ipynb`):
$$
\big\langle\Theta^{(p)}_{ij}(\mathbf r)\big\rangle=\hbar c\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}\Big[D^{(p)}_{ij}-\tfrac12\delta_{ij}\big(D^{(p)}_{kk}+\kappa^2g^{(p)}_s\big)\Big],\qquad
f_i=-\sum_{p={\rm TM,TE}}\oint_C\big\langle\Theta^{(p)}_{ik}\big\rangle n_k\,dl ,
$$
для $z$-однородного идеального проводника. Для материала с $\varepsilon(\omega),\mu(\omega)$ поляризации связаны, и эту формулу заменяет тензорная функция Грина (§1), см. `Casimir_cylinders_Maxwell_general.ipynb`.
2. **П1 пройдена тремя способами:** изображения ($10^{-6}$), Лифшиц ($10^{-10}$), панели ($0{,}37\%$ при $h=0{,}02a$, линейно по $h$).
3. Попутно: у **отдельной** поляризации $\Theta_{xx}$ расходится у стенки, и знак расходимости у Дирихле и Неймана противоположный. Локальное давление вдоль граней надо считать по сумме ТМ+ТЕ, а не по каждой отдельно.

**Не проверено здесь:** нормировка $\langle H_iH_k\rangle$ для неидеального металла (связь ТМ–ТЕ); это проверка П3 (Лифшиц для $\varepsilon(\omega)$), следующий шаг.
