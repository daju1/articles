Нормальная сила на каждую пластину раскладывается на две составляющие: поперёк уголка (компенсируется второй пластиной) и вдоль оси уголка (складывается). F_{\text{тяги}}=-\frac{\pi^2\hbar c\,b}{1440}\,\frac{\cos2\alpha}{\cos^4\alpha}\left(\frac{1}{L_{\min}^3}-\frac{1}{L_{\max}^3}\right),\qquad 0<\alpha<\frac\pi4 ,

где $b$ — длина уголка «вглубь», $L_{\min}, L_{\max}$ — расстояния между сторонами на уровнях $z_{\min}, z_{\max}$ его рис. 7. Давление на элемент поверхности берётся из формулы для параллельных пластин $-\pi^2\hbar c/240\,d^4$ (PFA). Оценка Антипина: $F_{\text{тяги}}\,[\text{дин}]\approx217\,b\,[\text{см}]/L_{\min}^3\,[\text{нм}]$.

**Что в этом рассуждении — модель, а не следствие уравнений:**
1. давление считается только на *внутренних* гранях и только по формуле плоских пластин;
2. давление поля на *внешние* грани, торцы и область вершины не учитывается;
3. «импульс виртуального фотона» передаётся только по нормали к элементу — это верно для отражения от плоскости, но в уголке поле не является суммой плоских волн, отражённых от каждой грани по отдельности.

## 1. Геометрия задачи

![](antipin_corner_geometry.png)

Тело однородно вдоль оси $z$ на длине $b$; $b$ много больше всех поперечных размеров, торцы по $z$ не учитываются (двумерная задача; эффекты торцов — отдельный пункт §8). Все силы ниже — **на единицу длины** вдоль $z$: $\mathbf f=\mathbf F/b$.

Сечение — плоскость $(x,y)$. Ось $y$ — ось симметрии уголка, направлена **от вершины к раствору**. (У Антипина эта ось названа $z$, а «вглубь рисунка» — $b$; у нас $z$ — образующая, как в MenDrive.)

| обозначение | смысл |
|---|---|
| $\alpha$ | половина угла раствора: угол между осью $y$ и срединной линией крыла |
| $t$ | толщина крыла |
| $\ell$ | длина крыла вдоль срединной линии |
| $r_0,\ r_1=r_0+\ell$ | расстояния от точки пересечения срединных линий до концов крыла (вариант A) |
| $L(y)=2y\tan\alpha-t/\cos\alpha$ | ширина щели между внутренними гранями на высоте $y$; $L_{\min}=L(y_{\min})$, $L_{\max}=L(y_{\max})$ |
| $\rho_{\rm in},\ \rho_{\rm out},\ \rho_{\rm tip}$ | радиусы скругления дна канавки, внешнего угла вершины, кромок свободных торцов (вариант B) |
| $D$ | область, занятая телом (телами); $\partial D$ — её граница; $s$ — длина дуги вдоль $\partial D$ |
| $\mathbf n$ | внешняя нормаль; $\boldsymbol\tau=\hat{\mathbf z}\times\mathbf n$ — касательная |

**Вариант A** — два отдельных крыла, щель у вершины $L_{\min}>0$. Модель статьи Антипина ($L_{\min}$ — «уровень отсечки»).

**Вариант B** — одно тело с замкнутой вершиной (как в патенте). Конечные $\rho_{\rm in},\rho_{\rm out},\rho_{\rm tip}$ — «технологический предел»; ответ должен исследоваться в пределе $\rho\to0$.

Симметрия $x\to-x$ даёт $f_x=0$ для любого из вариантов. **Искомая величина — $f_y$.**

## 2. Физическая модель

1. Электромагнитное поле в вакууме вне тел и в материале тел; другие поля (электроны как отдельная система, фононы) не рассматриваются.
2. Материал — линейный, локальный, с проницаемостями $\varepsilon(\omega)$, $\mu(\omega)$, аналитическими в верхней полуплоскости $\omega$ (причинность). Основной случай — **изотропный и взаимный**. Частные случаи:
   - идеальный проводник: $\varepsilon\to\infty$ (условия $\mathbf E_t=0,\ B_n=0$ на $\partial D$);
   - плазма: $\varepsilon(\omega)=1-\omega_p^2/\omega^2$;
   - Друде: $\varepsilon(\omega)=1-\omega_p^2/\omega(\omega+i\gamma)$;
   - **невзаимный (гиротропный) материал** — отдельный случай, см. §2a.
3. Система находится в **термодинамическом равновесии** при температуре $T$ (основной случай $T=0$). Тела неподвижны, внешних источников нет.
4. Вопрос о силе ставится как вопрос о **среднем значении** силы в этом состоянии.

Предпосылки 3–4 существенны: при их нарушении (неравновесие, движение, нестационарная накачка) задача другая — см. §5(г).

## 2a. Невзаимный материал (ферромагнетик, намагниченная плазма)

Если часть уголка (внутренняя или внешняя) выполнена из ферромагнетика с остаточной намагниченностью $\mathbf M_r$, то:

- **Линейность сохраняется.** Гистерезис — нелинейность при большом сигнале. Вакуумные флуктуации — малые отклонения вокруг рабочей точки на петле $B(H)$, и для них отклик линеен с обратимой проницаемостью. Основная энергия флуктуаций при зазорах $\sim100$ нм лежит при $\omega\sim c/a\sim10^{15}$ с$^{-1}$, где $\mu(\omega)\to1$ (грубая оценка, без расчёта спектра).
- **Нарушается взаимность.** Отклик гиротропный (тензор Полдера, как в MenDrive):
$$
\hat\mu(\omega)=\begin{pmatrix}\mu & i\kappa & 0\\ -i\kappa & \mu & 0\\ 0&0&\mu_\parallel\end{pmatrix},\qquad \kappa\propto M_r,\qquad \hat\mu^{T}\neq\hat\mu .
$$
Аналогично для намагниченной плазмы — гиротропный $\hat\varepsilon(\omega)$. Функция Грина несимметрична: $G_{ij}(\mathbf r,\mathbf r')\neq G_{ji}(\mathbf r',\mathbf r)$.

**Что известно в равновесии** (Gelbwaser-Klimovsky, Graham, Kardar, Krüger, PRB 106, 115106 (2022), `ref/GelbwaserKlimovsky2022_…pdf`). Вывод начинается не с энергии, а с силы: флуктуационно-диссипационная теорема, локально в каждом теле; сила раскладывается по источникам — в остальных телах, в самом теле, в окружении. При равных температурах сумма точно равна градиенту
$$
\mathbf F_{\rm eq}(1)=-\nabla_{O_1}\mathcal F,\qquad
\mathcal F=\frac{\hbar}{\pi}\int_0^\infty d\omega\Big[\frac{1}{e^{\hbar\omega/k_BT}-1}+\frac12\Big]\,\mathrm{Im\,Tr}\,\log\big(1-\mathbb G_0\mathbb T_1\mathbb G_0\mathbb T_{\bar1}\big),
$$
той же формы, что для взаимных тел. Отсюда для уголка:

| | следствие |
|---|---|
| вариант B, одно тело | других тел нет, $\mathbb T_{\bar1}=0$, $\log1=0$: $\mathbf f=0$ при любой форме |
| вариант A, два крыла | $\mathbf F^{(1)}=-\mathbf F^{(2)}$ независимо от взаимности: сумма равна нулю |
| силы между крыльями | в главном порядке по отражениям $\mathcal F$ — сумма вкладов взаимных и антивзаимных частей $\mathbb T$; поправки линейной по $\kappa$ нет; для зеркальных тел антивзаимный вклад даёт **отталкивание** — гиротропия может ослабить сжатие крыльев (проверить для конкретной ориентации $\mathbf M_r$) |

**Оговорки:** ФДТ в каждом теле предполагает больцмановское распределение — замороженная $\mathbf M_r$ это квазиравновесие; перемагничивание или релаксация $\mathbf M_r$ выводит систему из равновесия.

**Вне равновесия** (тело теплее окружения): вращающий момент возможен даже в вакууме, самодвижение — только в высших порядках (Milton и др., PRA 108, 022809 (2023)). Это расход энергии на тепловое излучение, а не тяга от вакуума.

**Спорные работы:** латеральная сила у анизотропной поглощающей пластины (Nefedov, Rubi, Sci. Rep. 7, 6155 (2017)) и предложение измерять равновесный незатухающий тепловой ток по броуновскому смещению частиц (Khandekar, Jacob, NJP 21, 103030 (2019)). Gelbwaser-Klimovsky и др. ставят оба результата под сомнение: в трансляционно-инвариантной системе $\mathcal F$ не зависит от координаты вдоль пластины, значит равновесная латеральная сила равна нулю.


## 3. Уравнения поля

Записываем пару уравнений Максвелла в той же форме, что в MenDrive (гауссова система):
$$
\mathrm{rot}\,\vec E=\frac{i\,\omega}{c}\,\mu\,\vec H,\qquad
\mathrm{rot}\,\vec H=-\frac{i\,\omega}{c}\,\varepsilon\,\vec E ,
$$
ищем решение в виде, позволяющем разделить переменные:
$$
\vec E(t)=\vec E(x,y)\,e^{i\,k_z\,z}\,e^{-i\,\omega\,t},\qquad \vec H(t)=\vec H(x,y)\,e^{i\,k_z\,z}\,e^{-i\,\omega\,t}.
$$
В каждой однородной области $\gamma^2=\varepsilon\mu\,\omega^2/c^2-k_z^2$, все компоненты выражаются через $E_z,H_z$:
$$
\vec E_\perp=\frac{i}{\gamma^2}\Big(k_z\nabla_\perp E_z-\frac{\omega}{c}\mu\,\hat{\mathbf z}\times\nabla_\perp H_z\Big),\qquad
\vec H_\perp=\frac{i}{\gamma^2}\Big(k_z\nabla_\perp H_z+\frac{\omega}{c}\varepsilon\,\hat{\mathbf z}\times\nabla_\perp E_z\Big),
$$
$$
(\nabla_\perp^2+\gamma^2)E_z=0,\qquad(\nabla_\perp^2+\gamma^2)H_z=0 .
$$
(Вывод и символьная проверка — `Casimir_cylinders_Maxwell_general.ipynb`, §1.)

**Граничные условия на $\partial D$:** непрерывны $E_z,\ H_z,\ E_s=\boldsymbol\tau\cdot\vec E,\ H_s=\boldsymbol\tau\cdot\vec H$. При $k_z\neq0$ и $\varepsilon_{\rm in}\neq\varepsilon_{\rm out}$ условия для $E_s,H_s$ **связывают** поляризации ТМ ($E_z$) и ТЕ ($H_z$); для идеального проводника они развязываются: $E_z|_{\partial D}=0$, $\partial_nH_z|_{\partial D}=0$.

## 4. Определение силы

**Тензор напряжений Максвелла** в вакууме ($\varepsilon=\mu=1$):
$$
T_{ik}=\frac{1}{4\pi}\Big[E_iE_k+H_iH_k-\tfrac12\delta_{ik}\big(E^2+H^2\big)\Big].
$$
**Сила на тело** (на единицу длины), контур $C$ целиком в вакууме, охватывает тело $D$ и не охватывает других тел, $\mathbf n$ — внешняя нормаль к $C$:
$$
f_i=\oint_C\big\langle T_{ik}\big\rangle\,n_k\,dl ,\qquad i,k\in\{x,y\}.
$$
Это — **определение** измеримой силы: импульс, который поле передаёт всему, что находится внутри $C$. Оно не опирается ни на энергию, ни на принцип виртуальной работы.

Для варианта A кроме полной силы на уголок имеет смысл и сила $\mathbf f^{(1)},\mathbf f^{(2)}$ на каждое крыло (контур вокруг одного крыла, не задевающий другое). Для варианта B — только полная сила.

**Вакуумное среднее** $\langle E_iE_k\rangle$, $\langle H_iH_k\rangle$ выражается через запаздывающую функцию Грина системы (флуктуационно-диссипационная теорема; Лифшиц–Питаевский, т. IX). После поворота контура частот $\omega=i\xi$ при $T=0$
$$
\big\langle T_{ik}(\mathbf r)\big\rangle=\frac{\hbar}{\pi}\int_0^\infty d\xi\;\mathcal T_{ik}(\mathbf r;\xi),\qquad
\text{при }T>0:\ \ \frac{\hbar}{\pi}\int_0^\infty d\xi\;\to\;2k_BT\sum_{n=0}^{\infty}{}'\Big|_{\xi_n=2\pi nk_BT/\hbar},
$$
где $\mathcal T_{ik}$ строится из **рассеянной** части тензорной функции Грина в совпадающих точках (свободная часть даёт однородный вклад с нулевой дивергенцией и вычитается). Нормировочные множители выведены и проверены в `Casimir_T_normalization_P1.ipynb`. Для $z$-однородного идеального проводника (поляризации $p=$ ТМ, ТЕ — скаляры Дирихле и Неймана):
$$
\big\langle\Theta^{(p)}_{ij}\big\rangle=\hbar c\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}\Big[D^{(p)}_{ij}-\tfrac12\delta_{ij}\big(D^{(p)}_{kk}+\kappa^2g^{(p)}_s\big)\Big],\qquad
f_i=-\sum_p\oint_C\big\langle\Theta^{(p)}_{ik}\big\rangle n_k\,dl ,
$$
$D_{ij}=\partial_i\partial'_jg_s|_{\mathbf r'=\mathbf r}$, $\kappa^2=\xi^2/c^2+k_z^2$. П1 пройдена тремя способами (изображения, Лифшиц, панели).

## 5. Что следует из постановки до всякого расчёта

**(а) Независимость от контура.** В равновесном статическом состоянии $\partial_k\langle T_{ik}\rangle=0$ всюду в вакууме. По теореме Гаусса $f_i$ одинакова для любых двух контуров, между которыми нет вещества: от большой окружности до контура, прижатого к поверхности тела на расстояние $\varepsilon\to0$. Скачок тензора на поверхности металла охватывается любым таким контуром целиком.

**(б) Нелокализуемость.** Сила на *часть* поверхности (на одну грань, торец, дно канавки) не определяется однозначно: у кромок $\langle T_{ik}\rangle$ сингулярен, и вклады соседних частей перетекают друг в друга. Для варианта A с двумя телами определён *тензор взаимодействия* $\langle T_{ik}\rangle[1{+}2]-\langle T_{ik}\rangle[1]-\langle T_{ik}\rangle[2]$, давление которого на грани конечно; для варианта B такого разбиения нет.

**(в) Две гипотезы.**
- $H_0$: $f_y=0$. Довод: свободная энергия равновесной системы не меняется при сдвиге всего тела в однородном вакууме.
- $H_1$: $f_y\neq0$ (Антипин).

Довод за $H_0$ использует энергию и потому сам по себе здесь не принимается как доказательство. Задача — **различить $H_0$ и $H_1$ прямым расчётом потока тензора** (§4), независимым от энергии.

**(г) Что могло бы дать $H_1$, если $H_0$ верна в статике.** Нарушение любой из предпосылок §2: неравновесие ($T_{\rm тела}\neq T_{\rm окружения}$), движение или переменные во времени границы, нелинейные материалы. Такие случаи — отдельная задача. **Невзаимность сама по себе** $H_1$ в равновесии не даёт (§2a).
