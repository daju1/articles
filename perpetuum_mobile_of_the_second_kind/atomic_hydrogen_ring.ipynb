Поскольку описанная в упомянутых источниках СПМ - симметричная (ванадий-палладиевый сплав, покрытый микронными слоями палладия и с входной, и с выходной стороны), эффекты автоматического сжатия проникающего водорода на порядки величины и мембранная рециркуляция водорода, при которой сжатый мембраной водород возвращается обратно во входную камеру, не могут быть обусловлены ничем иным, кроме резкой асимметрии начальных условий, при которых находится водород по разные стороны симметричной мембраны.

Ведь очевидно же, что - будь эти условия симметричны - в этой системе обязательно проявился бы прямой осмос, формирующий поток водорода противоположного направления: из камеры с чистым водородом в камеру со смесью газов, из которых мембрана проницаема лишь для одного - водорода. Однако фактически имеет место обратный осмос (молекулярная фильтрация одного газа), который, как известно - процесс принудительный, требующий совершения работы по продавливанию смеси газов через мембрану против сил осмотического давления. Представляется, что асимметричным фактором в окружении симметричной СПМ, настолько сильным, что он не только преодолевает естественные силы осмотического давления (о которых, на мой взгляд, забывать нельзя), но и водород в выходной камере сжимается на порядки величины, здесь является то, что водород во входной камере находится в состоянии плазмы, в которой его частицы гораздо более горячие, чем водородные частицы в выходной камере. 

Пока придерживаюсь мнения, что описанная структурная асимметрия мембраны заменит асимметрию ее водородного окружения, и она в динамике проявит те же эффекты автоматического сжатия водорода (очевидно, что не на порядки величины - но этого и не требуется), и его мембранной рециркуляции, причем с возможностью совершения работы (!).

Действительно, молекулярный водород во входной камере, благодаря каталитическим свойствам обращенного туда слоя Pd, у входной поверхности мембраны диссоциирует на атомы, диффундирующие вглубь мембраны к ее выходной поверхности. Далее они проникают в выходную камеру, преодолевая проницаемый для атомарного водорода микронный слой V. Поскольку атомарный водород неустойчив и, при возможности (отсутствующей, пока он находится в междоузлиях кристаллической решетки мембраны), ассоциирует в молекулярный, за выходной поверхностью мембраны оказывается молекулярный водород, не способный проникнуть вглубь. Обратно диссоциировать в атомарный ему не позволяет барьерный слой не обладающего каталитическими свойствами V, отделяющий водород в выходной камере от вышедших на поверхность мембраны атомов каталитического Pd, входящего в состав ее сплава.
