Гомогенный осмос уникален тем, что в нем относительно мелкие (проходящие сквозь поры в мембране) частицы растворителя, и крупные (задерживаемые мембраной) частицы растворенного вещества вне пределов мембраны не сохраняют постоянство, а способны, в зависимости от внешних условий (в частности, при наличии катализатора или при отсутствии пространственных ограничений), естественным образом претерпевать взаимопревращения.

Если это признать, то единственный фактор, не позволяющий сделать известные типы осмотических двигателей "вечными" - выравнивание в результате осмоса концентраций растворов по разные стороны от мембраны, обнуляющее осмотическое давление - в гомогенных растворах теряет силу, поскольку в них движущая разность концентраций при циркуляции раствора может поддерживаться извне, например, асимметричными граничными условиями системы. Что, если не это, предопределяет возможность сделать гомогенную осмотическую систему устойчиво-неравновесной, т.е, по Жвирблису, - кольцаром? Понимание этих процессов имеет важное значение для разработки новых метаматериалов, сообщила пресс-служба МФТИ.

    Мы выяснили, при каких условиях невзаимные системы могут обеспечивать перенос тепла от более холодной области к более горячей, а также приводить к экспоненциальной локализации высоких температур на границе системы. Результаты работы могут быть использованы для создания микроскопических тепловых машин, а также нового класса сенсоров

- пояснил научный сотрудник МФТИ (Долгопрудный) Даниил Колотинский, чьи слова приводит пресс-служба вуза.

Исследователи совершили это открытие при изучении эффектов, возникающих внутри упорядоченных плазменно-пылевых структур. В этих системах пылевые частицы взаимодействуют друг с другом, плазмой и нейтральным газом, однако кинетическая энергия, как обнаружили ученые, может быть очень неравномерно распределена по пылевой компоненте.

    Мы обнаружили, что в неравновесной пылевой плазме могут формироваться нетривиальные распределения кинетической энергии пылевых частиц, которые наблюдаются даже при постоянной и однородной температуре окружающего газа. Данный факт выходит за рамки классических представлений о термодинамическом равновесии и теореме о равнораспределении энергии по степеням свободы

- добавил доцент МФТИ Алексей Тимофеев, чьи слова приводит пресс-служба вуза.

Данные особенности поведения скоплений пыли и плазмы, а также обнаруженные учеными возможные нарушения второго начала термодинамики, заставили их создать математическую модель, которая описывает поведение частиц внутри упорядоченных плазменно-пылевых структур. Используя эту модель, построенную на базе так называемой неэрмитовой теории, ученые просчитали, как протекает процесс переноса тепла в системе и как на него влияют различные эффекты, возникающие из-за асимметричности взаимодействий.

    Предложенная нами модель позволяет объяснить происхождение и особенности этого явления.
