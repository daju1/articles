# «Преобразователь Лаврентьева»: даёт ли градиентный дрейф ток при одной температуре?

**Источник:** О.А. Лаврентьев, *Экспериментальные доказательства возможности преобразования тепловой энергии хаотического движения частиц непосредственно в электрическую*, препринт ХФТИ 92-24 (Харьков, 1992), 16 с. Скан без текстового слоя, все формулы ниже **прочитаны с изображения страниц**. Мелкие множители (π, 2) в отдельных формулах читаются по скану ненадёжно, где так, это отмечено.

**Рабочий критерий:** ток при равных термодинамических силах (одни T и μ) возможен только при нарушении условия Вегшайдера по циклу / детального равновесия / гиббсовости термостата.

**Содержание ноутбука**
1. Что утверждает препринт (ф. 1–25, схема, эксперимент) и проверка ф. (1)–(2) по рис. 3.
2. Аналитика: максвелл-больцмановское распределение стационарно для уравнения Власова с любым B (и с любым электростатическим потенциалом); граничные условия (A) с ним совместимы, а вариант (B) с поглощающими неэмитирующими внешними сторонами несовместим; дрейф ∇B в точности компенсируется током намагниченности.
3. Монте-Карло: 2D, электроны в B_z(y), дрейф B×∇B направлен поперёк зазора, интегратор Бориса, термализующие столкновения. Варианты ГУ: (A) детальное равновесие на всех сторонах; (B-refl) эмиссия только с внутренних сторон, внешние зеркально отражают; (B-abs) эмиссия только с внутренних сторон, внешние поглощают; (L) идеализация автора, где пластины перекрывают всю область градиента.
4. Оценки экспериментальных артефактов: ΔT катодов по Ричардсону–Дешману, эдс Нернста.

Все код-ячейки используют только стандартную библиотеку Python (без numpy/scipy). Каждая проверка печатает `OK`/`FAIL`.

## 1. Что в препринте (пересказ по скану)

**Схема (рис. 1).** Две параллельные плоские пластины A и B (зазор d) из одного материала, при одной температуре T, соединены через нагрузку R. Поле **B** параллельно пластинам, **∇B** тоже параллелен пластинам и перпендикулярен **B**, поэтому дрейф **B×∇B** направлен *от пластины к пластине*. Ключевое допущение (с. 2): плотность тока термоэмиссии j_s одинакова **на внутренних** поверхностях, а **на внешних поверхностях** она «принимается равной нулю».

**Формулы (гауссова система):**
* (1) v_d0 = ½ v_⊥ ρ_e |∇B|/B = (m v_⊥²/2e)·c|∇B|/B²: стандартный ∇B-дрейф ведущего центра. По мнению автора, именно он создаёт ток в нагрузке.
* (2) с тормозящим полем E = IR/d: v_d = (m v_⊥²/2e)(c|∇B|/B²)(1 − πcE/(v_⊥B)). Рис. 3 сравнивает это с численными траекториями (рис. 2).
* (3)–(5) угол отрыва от поверхности: α ≤ arccos(1 − 2πv_d/v_⊥) ≈ (4πv_d/v_⊥)^{1/2}. (6) j_k/j_s: интеграл по полумаксвеллу «оторвавшихся» электронов.
* (7)–(10) пространственный заряд приэлектродного слоя толщиной 2ρ_e: n_e = πj_s/(2v_⊥), U = 4π²j_sρ_e/ω_ce, E_ср = 2π²j_s/ω_ce, x_m = 2ρ_e − 2eE/(mω_ce²). (11) j_k/j_s = A∫_{b²}^∞ e^{−t}(√t − b)^{1/2}dt (рис. 5: при больших j_s ток инжекции «запирается»).
* (12)–(14) пролётная скорость и ток через диод, ограниченный объёмным зарядом (рис. 6).
* (15)–(21) учёт столкновений с водородом: диффузионное заполнение промежутка, больцмановское распределение n_e = n_e0/cos²(…) (это решение (17)–(19)), n_e0 из условия eU(0) = kT_e.
* (22)–(24) дрейфовый ток при столкновительном заполнении: j_0 = n_e0 c kT_e|∇B|/B²·[1 − (b+1)e^{−b}], b = m ω_ce² d²/(8kT_e); (23) j_D − j_M + j = const; (25) диффузионный ток. Табл. 1: T_e = 1400 K, n_a = 10¹⁴ см⁻³ (H₂).

**Эксперимент (с. 13–15):** термоэмиссионные шайбы диаметром 5 мм (по скану читается «из гексаборида лантана», запрессованы в танталовые стаканы) в стеклянной колбе с водородом, p = 1.8·10⁻⁴ мм рт. ст. (n_a = 1.5·10¹³ см⁻³). Подогреватели питаются от отдельных накальных трансформаторов. Цена деления гальванометра 1.5·10⁻⁹ А. Неоднородное поле создаёт постоянный магнит, который отодвигают от диода. **Процедура:** задают одинаковую мощность накала, затем *тонкой подстройкой накала уравнивают токи эмиссии левого и правого катодов, то есть выставляют ноль гальванометра при B = 0*, и только после этого накладывают поле. Температуры на рис. 10: **1000 ± 100, 1200 ± 100, 1400 ± 100 K**. Ток ограничен пространственным зарядом и составляет **≈2·10⁻⁸ А** (на 1300 Ом это ≈5·10⁻¹³ Вт; это согласуется: (2·10⁻⁸)²·1300 = 5.2·10⁻¹³). При смене полярности поля ток меняет знак.

**Где ошибка в теории.** Скорость (1) описывает дрейф *ведущего центра*, а не поток частиц. В равновесном газе поток частиц через любое сечение равен нулю: дрейфовый поток в точности гасится током намагниченности c∇×M, который на границах проявляется как скачущие орбиты (раздел 2). Ток появляется у автора только за счёт ГУ «внешние стороны не эмитируют», а это прямое нарушение детального равновесия на поверхности (раздел 3).

```sage
import math, random, time
RESULTS = []
def check(name, cond, info=""):
    s = "OK  " if cond else "FAIL"
    RESULTS.append((name, bool(cond)))
    print("[%s] %s %s" % (s, name, ("| " + info) if info else ""))
    return bool(cond)

# --- Проверка ф. (1),(2) препринта по данным рис. 3 (гауссова система) ---
me, e, c = 9.109e-28, 4.803e-10, 2.998e10      # г, СГСЭ, см/с
v_perp, B, gradB = 1.45e7, 10.0, 10.0           # параметры рис.3: v=1.45e7 см/с, B=10 Гс, |grad B|=10 Гс/см
vd0 = me*v_perp**2/(2*e) * c*gradB/B**2         # ф.(1)
E0_stat = v_perp*B/(math.pi*c)                   # нуль ф.(2): 1 - pi c E/(v B) = 0
E0_Vcm = E0_stat*299.79
print("ф.(1): v_d0 = %.3g см/с  (рис.3, |grad B|=10 Гс/см: при E=0 ~6·10^5 см/с, если ось в ед. 10^5 см/с)" % vd0)
print("ф.(2): v_d = 0 при E = %.3f В/см (на рис.3 прямые пересекают ось около 0.45-0.5 В/см)" % E0_Vcm)
check("ф.(1) воспроизводит отсечку рис.3 (5..7)e5 см/с", 5e5 < vd0 < 7e5, "%.3g" % vd0)
check("ф.(2) даёт нуль дрейфа в 0.40..0.55 В/см как на рис.3", 0.40 < E0_Vcm < 0.55, "%.3f В/см" % E0_Vcm)
# Ларморовский радиус и частоты в эксперименте (для сопоставления с безразмерной моделью)
kB_erg = 1.380649e-16; T = 1200.0
vth = math.sqrt(kB_erg*T/me)
for Bg in (1.0, 10.0, 71.0):
    wc = e*Bg/(me*c)
    print("B=%5.1f Гс: v_th=%.3g см/с, omega_c=%.3g 1/с, rho_th=%.3g см" % (Bg, vth, wc, vth/wc))
na, sigma = 1.5e13, 1e-15                       # H2, порядок сечения ~1e-15 см^2 (оценка)
print("частота столкновений nu ~ n_a sigma v_th = %.2g 1/с;  nu/omega_c(10 Гс) ~ %.1g" % (na*sigma*vth, na*sigma*vth/(e*10/(me*c))))
```

## 2. Аналитика: стационарность Максвелла–Больцмана и граничные условия

Стационарное уравнение Власова с термализующим интегралом столкновений (BGK) в 2D (B = B_z(x,y) ẑ, произвольный электростатический потенциал Φ(x,y)):

$$\mathbf v\cdot\nabla_{\mathbf r} f+\frac{q}{m}\big(-\nabla\Phi+\mathbf v\times\mathbf B\big)\cdot\nabla_{\mathbf v} f=\nu\,(f_{\rm eq}-f).$$

Для $f_{\rm eq}=n_0\,e^{-q\Phi/kT}\,\frac{m}{2\pi kT}e^{-mv^2/2kT}$ получаем $\nabla_{\mathbf v}f=-\frac{m\mathbf v}{kT}f$, $\nabla_{\mathbf r}f=-\frac{q\nabla\Phi}{kT}f$. Тогда
* член с полем E: $\mathbf v\cdot(-q\nabla\Phi/kT)f+\frac{q}{m}(-\nabla\Phi)\cdot(-\frac{m\mathbf v}{kT})f=0$;
* магнитный член: $(\mathbf v\times\mathbf B)\cdot\mathbf v=0$, **при любом B(r)**;
* столкновительный член равен нулю, потому что $f=f_{\rm eq}$.

Значит, гиббсово состояние стационарно при любой неоднородности B и при любом самосогласованном или контактном потенциале. Плотность тока $\mathbf j=q\int\mathbf v f\,d^2v=0$ *в каждой точке* (подынтегральное выражение нечётно по v). Отсюда теорема Бора–ван Лёвен: равновесный ток через любое сечение, в том числе между пластинами, равен нулю.

**Почему дрейф (1) не даёт тока.** Поток ведущих центров $j_{gc,x}=qn\langle v_{\nabla B,x}\rangle=-nkT\,B'/B^2$ (для B = B(y)ẑ, в 2D $\langle mv_\perp^2/2\rangle=kT$). Диамагнитная намагниченность $M_z=-nkT/B$ даёт ток $(\nabla\times M)_x=\partial_yM_z=+nkT\,B'/B^2$. Сумма равна нулю. Автор учитывает только первое слагаемое.

**Граничные условия.** На поверхности с нормалью n для уходящих скоростей (v·n > 0):
$f_{\rm out}(\mathbf v)=\epsilon\, f_M(\mathbf v)+(1-a)\,f_{\rm in}(R\mathbf v)$, где a — вероятность поглощения, ε — эмиссионная способность (эмитируется поток εΦ_eq с полумаксвелловским распределением), R — зеркальное отражение. Подстановка $f_{\rm in}=f_M$ даёт $f_{\rm out}=f_M$ **тогда и только тогда, когда ε = a** (закон Кирхгофа для электронов, то есть детальное равновесие на поверхности). Получается:
* (A) ε = a = α на обеих сторонах: совместимо, ток = 0;
* (B-refl) внешние стороны ε = 0, a = 0 (зеркало): совместимо, ток = 0;
* (B-abs) внешние стороны ε = 0, a = α > 0: **несовместимо**. Внешняя сторона ведёт себя как сток (резервуар с μ → −∞), гиббсовость нарушена граничным условием, и ток разрешён.

```sage
# --- 2a. Численная проверка (конечные разности): L[f] = 0 для f_eq при произвольных B_z(x,y), Phi(x,y) ---
q, m, kT, n0 = -1.0, 1.0, 1.0, 1.0
def Bz1(x, y): return 1.0 + 0.15*y
def Bz2(x, y): return 0.7 + 0.3*math.sin(1.3*y) + 0.2*x*y          # сильно неоднородное
def Phi0(x, y): return 0.0
def Phi1(x, y): return 0.4*math.cos(0.7*x) - 0.3*y*y                # «самосогласованный/контактный» потенциал
def f_eq(Phi):
    return lambda x, y, vx, vy: n0*math.exp(-q*Phi(x, y)/kT)*m/(2*math.pi*kT)*math.exp(-m*(vx*vx+vy*vy)/(2*kT))
def vlasov_residual(f, Bz, Phi, x, y, vx, vy, h=1e-5):
    fx  = (f(x+h,y,vx,vy)-f(x-h,y,vx,vy))/(2*h)
    fy  = (f(x,y+h,vx,vy)-f(x,y-h,vx,vy))/(2*h)
    fvx = (f(x,y,vx+h,vy)-f(x,y,vx-h,vy))/(2*h)
    fvy = (f(x,y,vx,vy+h)-f(x,y,vx,vy-h))/(2*h)
    Ex  = -(Phi(x+h,y)-Phi(x-h,y))/(2*h); Ey = -(Phi(x,y+h)-Phi(x,y-h))/(2*h)
    B = Bz(x, y)
    Fx = q*(Ex + vy*B); Fy = q*(Ey - vx*B)      # q(E + v x B), B = B_z z
    L = vx*fx + vy*fy + (Fx*fvx + Fy*fvy)/m
    scale = (abs(vx*fx)+abs(vy*fy)+abs(Fx*fvx/m)+abs(Fy*fvy/m)) + 1e-300
    return abs(L)/scale
rng = random.Random(2024)
pts = [(rng.uniform(-3,3), rng.uniform(0,6), rng.gauss(0,1.5), rng.gauss(0,1.5)) for _ in range(300)]
for (Bn, Bz) in (("B=1+0.15y", Bz1), ("B=0.7+0.3sin(1.3y)+0.2xy", Bz2)):
    for (Pn, Phi) in (("Phi=0", Phi0), ("Phi=0.4cos(0.7x)-0.3y^2", Phi1)):
        r = max(vlasov_residual(f_eq(Phi), Bz, Phi, *p) for p in pts)
        check("Власов: f_eq стационарно, %s, %s" % (Bn, Pn), r < 1e-6, "max отн. невязка %.1e" % r)
# контрпримеры: проверка не тривиальна
fgrad  = lambda x, y, vx, vy: (1+0.2*y)*f_eq(Phi0)(x, y, vx, vy)       # градиент плотности без поля E
fshift = lambda x, y, vx, vy: f_eq(Phi0)(x, y, vx-0.1, vy)             # «дрейфующий» максвелл
r1 = max(vlasov_residual(fgrad, Bz1, Phi0, *p) for p in pts)
r2 = max(vlasov_residual(fshift, Bz1, Phi0, *p) for p in pts)
check("контрпример n(y)*f_M без E НЕ стационарен", r1 > 1e-2, "max %.2g" % r1)
check("контрпример сдвинутый максвелл (поток) НЕ стационарен", r2 > 1e-2, "max %.2g" % r2)
```

```sage
# --- 2b. Граничные условия: f_out = eps*f_M + (1-a)*f_M(Rv) == f_M  <=>  eps == a ---
fM = lambda vx, vy: math.exp(-(vx*vx+vy*vy)/2)/(2*math.pi)
def bc_residual(eps, a, nsamp=200, seed=1):
    r = random.Random(seed); worst = 0.0
    for _ in range(nsamp):
        vn = abs(r.gauss(0,1.3)); vt = r.gauss(0,1.3)      # уходящая скорость (v.n>0)
        fout = eps*fM(vn, vt) + (1-a)*fM(-vn, vt)          # Rv = (-vn, vt) — падающая
        worst = max(worst, abs(fout - fM(vn, vt))/fM(vn, vt))
    return worst
alpha = 0.7
cases_bc = [("(A) обе стороны: eps=a=alpha", alpha, alpha, True),
            ("(B-refl) внешняя: eps=0, a=0 (зеркало)", 0.0, 0.0, True),
            ("(B-abs) внешняя: eps=0, a=alpha (сток)", 0.0, alpha, False)]
for name, eps, a, compatible in cases_bc:
    res = bc_residual(eps, a)
    if compatible:
        check("ГУ совместимо с f_M: " + name, res < 1e-12, "невязка %.1e" % res)
    else:
        check("ГУ НЕсовместимо с f_M (нарушено детальное равновесие): " + name, res > 0.1, "f_out/f_M - 1 = %.2f" % (-res))

# --- 2c. j = q * int v f_M d^2v = 0 и компенсация дрейфа ∇B током намагниченности ---
hv = 0.05; grid = [(-8 + hv*i) for i in range(int(16/hv)+1)]
jx = sum(vx*fM(vx, vy) for vx in grid for vy in grid)*hv*hv
Ekin = sum(0.5*(vx*vx+vy*vy)*fM(vx, vy) for vx in grid for vy in grid)*hv*hv
check("равновесный поток частиц j_x = 0", abs(jx) < 1e-12, "%.1e" % jx)
check("<m v_perp^2/2> = kT в 2D (квадратура)", abs(Ekin-1.0) < 1e-6, "%.8f" % Ekin)
def jgc_x(y, Bz=Bz1, n=1.0, h=1e-5):           # q n <v_gradB,x>,  v_gradB,x = -(m v^2/2) B'/(q B^2)
    Bp = (Bz(0,y+h)-Bz(0,y-h))/(2*h); B = Bz(0,y)
    return q*n*(-(Ekin*kT)*Bp/(q*B*B))
def jM_x(y, Bz=Bz1, n=1.0, h=1e-5):             # (curl M)_x = dM_z/dy, M_z = -n kT/B (диамагнетизм)
    Mz = lambda yy: -n*kT/Bz(0, yy)
    return (Mz(y+h)-Mz(y-h))/(2*h)
worst = max(abs(jgc_x(y)+jM_x(y)) for y in [0.3*i for i in range(1,20)])
print("пример y=2: j_gc,x = %.5f, (curl M)_x = %.5f" % (jgc_x(2.0), jM_x(2.0)))
check("дрейфовый поток ведущих центров + ток намагниченности = 0", worst < 1e-8, "max |сумма| %.1e" % worst)
```

## 3. Модель Монте-Карло

**Геометрия (безразмерные единицы: m = kT = 1, q = −1 (электрон), ω_c0 = |q|B_0/m = 1, ρ_th = 1).** Ящик x ∈ [−5, 5], y ∈ [0, 8] с зеркальными стенками (стекло колбы; зеркальная стенка совместима с равновесием). Пластины — бесконечно тонкие отрезки x = ∓d/2 (d = 3), y ∈ [2, 6] (высота h = 4). У каждой пластины две стороны: внутренняя (к зазору) и внешняя. Электроны могут обходить края пластин. Поле: $B_z(y)=B_0[1+g(y-4)]$, g = 0.15, то есть B от 0.4B₀ до 1.6B₀ (gρ = 0.15; в эксперименте gρ ~ 0.03, градиент в модели взят сильнее, чтобы эффекты были видны). Для q < 0, B₀ > 0 дрейф **B×∇B** направлен в +x, от пластины 1 к пластине 2.

**Динамика.** Схема leapfrog/Борис: поворот скорости в $B_z(y_n)$, затем столкновение (с вероятностью νΔt скорость заменяется выборкой из Максвелла при T), затем свободный пролёт с проверкой пересечения пластин и стенок. Δt = 0.1, ν = 0.1 (ν/ω_c = 0.1; в эксперименте ≈10⁻³, отдельно проверены ν = 0.02 и 0.5). Поворот Бориса точно сохраняет |v|, а отображение сохраняет фазовый объём, поэтому однородный максвелл инвариантен и в дискретном времени.

**Эмиссия** со стороны f: поток $\Gamma_f=\epsilon_f\,\Phi_{eq}h$, $\Phi_{eq}=n/\sqrt{2\pi}$ (2D). Нормальная компонента распределена по Рэлею (полумаксвелл, взвешенный потоком), касательная — по Гауссу. Частица ставится на долю u ∈ U(0,1) первого пролёта: это точная дискретная инжекция, иначе возникает систематика O(ωΔt), нечётная по B, и она была замечена в черновом прогоне. **Поглощение** с вероятностью a_f, иначе зеркальное отражение.

**Почему достаточно «трансмиссионного» МК.** Электроны без самосогласованного поля не взаимодействуют, поэтому стационарное состояние линейно по источникам. Стационарный ток 1→2 (в электронах в единицу времени):
$$J=\sum_{f\in 1}\Gamma_f P_{f\to 2}-\sum_{f\in 2}\Gamma_f P_{f\to 1},$$
где $P_{f\to k}$ — вероятность того, что частица, испущенная стороной f, будет поглощена пластиной k. Это строго эквивалентно прямому моделированию популяции. Плотность восстанавливается по времени пребывания (проверка в п. 3d). Ошибки биномиальные, $\sigma^2=\sum\Gamma_f^2P(1-P)/N$. Ниже J нормирован на $\Gamma_0=\Phi_{eq}h$, то есть на односторонний равновесный поток одной стороны при α = 1. Электрический ток равен $I=qJ=-eJ$.

```sage
def make_geom(case, B0, g=0.15, d=3.0, Lx=10.0, Ly=8.0, h=4.0, alpha=1.0):
    """case: 'A'    - все 4 стороны эмитируют и поглощают (eps=a=alpha);
             'Bref' - эмиссия+поглощение только внутренние; внешние - зеркало;
             'Babs' - эмиссия только внутренние; поглощают все 4 стороны;
             'L'    - идеализация автора: пластины перекрывают всю высоту области градиента."""
    G = dict(case=case, B0=B0, g=g, d=d, Ly=Ly, alpha=alpha, yc=Ly/2)
    if case == 'L': G.update(Lx=d, ya=0.0, yb=Ly)
    else:           G.update(Lx=Lx, ya=(Ly-h)/2, yb=(Ly+h)/2)
    a = alpha
    # стороны: 0 = пластина1 внутр., 1 = пластина1 внешн., 2 = пластина2 внутр., 3 = пластина2 внешн.
    if case == 'A':      G['absorb'] = [a, a, a, a]; G['emit'] = [a, a, a, a]
    elif case == 'Babs': G['absorb'] = [a, a, a, a]; G['emit'] = [a, 0, a, 0]
    elif case in ('Bref', 'L'): G['absorb'] = [a, 0, a, 0]; G['emit'] = [a, 0, a, 0]
    return G

def track(G, face, rng, dt=0.1, nu=0.1, q=-1.0, Tmax=3000.0, acc=None):
    """Траектория электрона, испущенного стороной face; возвращает (сторона поглощения или -1, время)."""
    d=G['d']; h2=d/2; Lx2=G['Lx']/2; Ly=G['Ly']; ya=G['ya']; yb=G['yb']
    B0=G['B0']; g=G['g']; yc=G['yc']; absorb=G['absorb']
    rnd = rng.random; gauss = rng.gauss
    xp = -h2 if face < 2 else h2
    sig = (1.0, -1.0, -1.0, 1.0)[face]           # нормаль стороны внутрь области, куда идёт эмиссия
    vn = math.sqrt(-2.0*math.log(1.0-rnd()))     # потоково-взвешенный полумаксвелл (Рэлей)
    vx = sig*vn; vy = gauss(0.0, 1.0)
    u = rnd()                                    # точная дискретная инжекция: доля первого пролёта
    x = xp + vx*u*dt
    y = ya + (yb-ya)*rnd() + vy*u*dt
    if y < 0.0: y = -y; vy = -vy
    elif y > Ly: y = 2*Ly - y; vy = -vy
    pc = nu*dt; hq = 0.5*q*dt
    for it in range(int(Tmax/dt)):
        tz = hq*B0*(1.0 + g*(y-yc))              # Борис: v_{n-1/2} -> v_{n+1/2} в B_z(y_n)
        vpx = vx + vy*tz; vpy = vy - vx*tz
        s = 2.0*tz/(1.0+tz*tz)
        vx = vx + vpy*s; vy = vy - vpx*s
        if rnd() < pc:                           # термализующее столкновение с газом
            vx = gauss(0.0,1.0); vy = gauss(0.0,1.0)
        if acc is not None: acc(x, y, vx, vy, dt)
        x1 = x + vx*dt; y1 = y + vy*dt           # свободный пролёт
        if (x + h2)*(x1 + h2) < 0.0: xp = -h2; k = 0
        elif (x - h2)*(x1 - h2) < 0.0: xp = h2; k = 2
        else: k = -1
        if k >= 0:
            yh = y + (y1-y)*(xp - x)/(x1 - x)
            if ya <= yh <= yb:                   # попали в пластину
                f = k + (0 if abs(x) < h2 else 1)
                af = absorb[f]
                if af > 0.0 and (af >= 1.0 or rnd() < af):
                    return f, it*dt
                x1 = 2*xp - x1; vx = -vx        # зеркальное отражение
        if x1 < -Lx2: x1 = -2*Lx2 - x1; vx = -vx
        elif x1 > Lx2: x1 = 2*Lx2 - x1; vx = -vx
        if y1 < 0.0: y1 = -y1; vy = -vy
        elif y1 > Ly: y1 = 2*Ly - y1; vy = -vy
        x = x1; y = y1
    return -1, Tmax

def run_case(case, B0, N, seed, **kw):
    G = make_geom(case, B0, **{k: v for k, v in kw.items() if k in ('g','d','Lx','Ly','h','alpha')})
    tkw = {k: v for k, v in kw.items() if k in ('dt','nu','Tmax')}
    rng = random.Random(seed); counts = {}
    for f in range(4):
        if G['emit'][f] <= 0: continue
        cnt = [0, 0, 0, 0, 0]                    # стороны 0..3, таймаут
        for _ in range(N):
            r, _t = track(G, f, rng, **tkw)
            cnt[r if r >= 0 else 4] += 1
        counts[f] = cnt
    return G, counts

def current(G, counts, N):
    """J/Gamma0 (электроны/ед.времени, 1->2) и стат. ошибка."""
    J = 0.0; var = 0.0
    for f, cnt in counts.items():
        Gf = G['emit'][f]
        if f < 2: p = (cnt[2]+cnt[3])/N; sgn = 1.0
        else:     p = (cnt[0]+cnt[1])/N; sgn = -1.0
        J += sgn*Gf*p; var += Gf*Gf*p*(1-p)/N
    return J, math.sqrt(var)
print("модель определена")
```

```sage
# --- 3a. Санитарные проверки интегратора ---
def free_orbit(B0, g, v=1.0, q=-1.0, dt=0.01, T=2000.0):
    x = 0.0; y = 0.0; vx = v; vy = 0.0; hq = 0.5*q*dt; e0 = vx*vx+vy*vy
    for it in range(int(T/dt)):
        tz = hq*B0*(1+g*y); vpx = vx+vy*tz; vpy = vy-vx*tz; s = 2*tz/(1+tz*tz); vx += vpy*s; vy -= vpx*s
        x += vx*dt; y += vy*dt
    return x/T, abs(vx*vx+vy*vy-e0)/e0
g_s = 0.02
vxd, de = free_orbit(1.0, g_s)
yg = 1.0                                  # ведущий центр: q<0, B>0, v=(1,0) -> центр при y=+rho=+1
vth_drift = (1.0**2/2)*g_s/(1+g_s*yg)**2  # v_x = -(m v^2/2) B'/(q B^2),  q=-1
check("Борис сохраняет |v| (2e5 шагов)", de < 1e-10, "%.1e" % de)
check("знак и величина ∇B-дрейфа: моделирование vs формула (1), 5%", abs(vxd/vth_drift-1) < 0.05,
      "v_x = %.5f, теория %.5f" % (vxd, vth_drift))
vxd_m, _ = free_orbit(-1.0, g_s)
check("при B -> -B дрейф меняет знак", vxd_m < 0 < vxd, "%.5f / %.5f" % (vxd, vxd_m))
```

### 3b. Основной эксперимент: ток между пластинами для B₀ = +1, 0, −1

Ожидания из раздела 2: (A), (B-refl), (L) дают J = 0 при любом B (гиббсово состояние совместимо с ГУ). (B-abs) не запрещён. Если ток есть, он нечётен по B: зеркало x → −x меняет местами пластины и переворачивает B_z.

```sage
N = 60000
RUNS = {}
t0 = time.time()
print("%-5s %5s %10s %8s %7s   %s" % ("ГУ", "B0", "J/Gamma0", "sigma", "J/sig", "вероятности (сторона испускания -> поглощения 0..3)"))
seed = 100
for case in ('A', 'Bref', 'Babs', 'L'):
    for B0 in (1.0, 0.0, -1.0):
        seed += 1
        G, cnt = run_case(case, B0, N, seed)
        J, s = current(G, cnt, N)
        RUNS[(case, B0)] = (G, cnt, J, s)
        pr = "; ".join("%d:%s" % (f, ",".join("%.3f" % (v/N) for v in cnt[:4])) for f, cnt in cnt.items())
        print("%-5s %+5.1f %+10.4f %8.4f %+7.2f   %s" % (case, B0, J, s, J/s, pr))
print("время %.0f с" % (time.time()-t0))
timeouts = sum(cnt[4] for (G, c, J, s) in RUNS.values() for cnt in c.values())
check("нет траекторий, оборванных по Tmax", timeouts == 0, str(timeouts))
for case in ('A', 'Bref', 'L'):
    for B0 in (1.0, 0.0, -1.0):
        J, s = RUNS[(case, B0)][2:]
        check("равновесное ГУ %s, B0=%+.0f: J = 0 (|J|<4 sigma)" % (case, B0), abs(J) < 4*s, "J=%+.4f±%.4f" % (J, s))
J0, s0 = RUNS[('Babs', 0.0)][2:]
check("B-abs, B0=0: J = 0 по симметрии", abs(J0) < 4*s0, "J=%+.4f±%.4f" % (J0, s0))
Jp, sp = RUNS[('Babs', 1.0)][2:]; Jm, sm = RUNS[('Babs', -1.0)][2:]
check("B-abs, B0=+1: J != 0 (>10 sigma)", abs(Jp) > 10*sp, "J=%+.4f±%.4f" % (Jp, sp))
check("B-abs, B0=-1: J != 0 (>10 sigma)", abs(Jm) > 10*sm, "J=%+.4f±%.4f" % (Jm, sm))
check("B-abs: ток нечётен по B, J(+B) = -J(-B)", abs(Jp+Jm) < 4*math.hypot(sp, sm), "сумма %+.4f±%.4f" % (Jp+Jm, math.hypot(sp, sm)))
print("НАБЛЮДЕНИЕ: в B-abs при B0>0 (дрейф ведущих центров +x, 1->2) поток электронов J = %+.3f Gamma0, "
      "т.е. направлен %s дрейфа." % (Jp, "ПРОТИВ" if Jp < 0 else "ПО"))
```

```sage
# --- 3c. Правила сумм и взаимность Онзагера в (A): тест детального равновесия на уровне P_{i->j} ---
NA = N
zs_sum = []
for B0 in (1.0, 0.0, -1.0):
    G, cnt, J, s = RUNS[('A', B0)]
    for k in range(4):           # поглощено стороной k = испущено стороной k  (все Gamma_f = 1)
        Pk = [cnt[f][k]/NA for f in range(4)]
        absorbed = sum(Pk); sig = math.sqrt(sum(p*(1-p)/NA for p in Pk))
        zs_sum.append((absorbed-1.0)/sig)
check("(A) правило сумм: поглощение = эмиссия на каждой стороне (12 тестов, max|z|<4.5)",
      max(abs(z) for z in zs_sum) < 4.5, "max|z| = %.2f" % max(abs(z) for z in zs_sum))
cp = RUNS[('A', 1.0)][1]; cm = RUNS[('A', -1.0)][1]
zs_ons = []
for i in range(4):
    for j in range(4):
        p1 = cp[i][j]/NA; p2 = cm[j][i]/NA          # Gamma_i P_{i->j}(B) = Gamma_j P_{j->i}(-B)
        sig = math.sqrt(p1*(1-p1)/NA + p2*(1-p2)/NA)
        zs_ons.append((p1-p2)/sig)
check("(A) взаимность Онзагера P_ij(B)=P_ji(-B) (16 тестов, max|z|<4.5)", max(abs(z) for z in zs_ons) < 4.5,
      "max|z| = %.2f" % max(abs(z) for z in zs_ons))
# прямой перенос внутр.->внутр. в (A) асимметричен, но компенсируется переносом через внешние стороны:
p02 = cp[0][2]/NA; p20 = cp[2][0]/NA
print("(A), B0=+1: P(внутр.1->внутр.2) = %.4f,  P(внутр.2->внутр.1) = %.4f  (асимметрия %+.4f)" % (p02, p20, p02-p20))
print("            остальные каналы компенсируют её: полный J = %+.4f ± %.4f" % RUNS[('A', 1.0)][2:])
```

```sage
# --- 3d. Восстановленная плотность и энергия: (A) должно дать однородное n=1, <E>=kT; (B-abs) - нет ---
def density_map(case, B0, Nf, seed, nxc=10, nyc=8):
    G = make_geom(case, B0); Lx, Ly = G['Lx'], G['Ly']
    tim = [[0.0]*nyc for _ in range(nxc)]; acc_e = [0.0, 0.0]
    rng = random.Random(seed); Phi = 1.0/math.sqrt(2*math.pi); h = G['yb']-G['ya']
    for f in range(4):
        if G['emit'][f] <= 0: continue
        w = G['emit'][f]*Phi*h/Nf
        def acc(x, y, vx, vy, dt, w=w):
            i = min(int((x+Lx/2)/Lx*nxc), nxc-1); j = min(int(y/Ly*nyc), nyc-1)
            tim[i][j] += w*dt; acc_e[0] += w*dt*(vx*vx+vy*vy)/2; acc_e[1] += w*dt
        for _ in range(Nf): track(G, f, rng, acc=acc)
    area = (Lx/nxc)*(Ly/nyc)
    return [[tim[i][j]/area for j in range(nyc)] for i in range(nxc)], acc_e[0]/acc_e[1], acc_e[1]/(Lx*Ly)
t0 = time.time()
dA, EA, nA = density_map('A', 1.0, 20000, 7)
flat = [v for row in dA for v in row]
print("(A), B0=+1: плотность по ячейкам 1x1 (строки - y сверху вниз):")
for j in range(7, -1, -1): print("   " + " ".join("%.2f" % dA[i][j] for i in range(10)))
check("(A) средняя плотность n = 1 (2%)", abs(nA-1) < 0.02, "%.4f" % nA)
check("(A) однородность: все ячейки в 1±0.1", max(abs(v-1) for v in flat) < 0.1, "min %.3f, max %.3f" % (min(flat), max(flat)))
check("(A) средняя кин. энергия = kT (2%)", abs(EA-1) < 0.02, "%.4f" % EA)
dB, EB, nB = density_map('Babs', 1.0, 20000, 8)
outer = [dB[i][j] for i in (0, 1, 8, 9) for j in range(8)]; gap = [dB[i][j] for i in (4, 5) for j in range(2, 6)]
print("(B-abs), B0=+1: <n> = %.3f, n(зазор) = %.3f, n(внешние области) = %.3f, <E> = %.3f" %
      (nB, sum(gap)/len(gap), sum(outer)/len(outer), EB))
check("(B-abs) состояние НЕ гиббсово (плотность обеднена стоками, <n> < 0.9)", nB < 0.9, "%.3f" % nB)
print("время %.0f с" % (time.time()-t0))
```

```sage
# --- 3e. B-abs: зависимость от B0 и устойчивость к dt, nu, alpha ---
Ns = 20000
print("B-abs: J/Gamma0 vs B0")
zodd = []; scan = {}
for B0 in (0.25, 0.5, 1.0, 2.0):
    Jp_, sp_ = current(*run_case('Babs',  B0, Ns, 300+int(10*B0)), Ns)
    Jm_, sm_ = current(*run_case('Babs', -B0, Ns, 400+int(10*B0)), Ns)
    scan[B0] = (Jp_, sp_, Jm_, sm_); zodd.append((Jp_+Jm_)/math.hypot(sp_, sm_))
    print("  B0=%4.2f: J(+B)=%+.4f±%.4f   J(-B)=%+.4f±%.4f" % (B0, Jp_, sp_, Jm_, sm_))
check("B-abs: нечётность по B на всём скане (max|z|<4)", max(abs(z) for z in zodd) < 4, "max|z|=%.2f" % max(abs(z) for z in zodd))
print("B-abs, B0=+1: чувствительность к параметрам")
for lab, kw in (("dt=0.05", {'dt': 0.05}), ("nu=0.02", {'nu': 0.02}), ("nu=0.5", {'nu': 0.5}), ("alpha=0.5", {'alpha': 0.5})):
    G_, c_ = run_case('Babs', 1.0, Ns, 500+len(lab), **kw); J_, s_ = current(G_, c_, Ns)
    print("  %-9s J=%+.4f±%.4f" % (lab, J_, s_))
    check("B-abs, %s: ток сохраняет знак и значим (>5 sigma)" % lab, J_ < 0 and abs(J_) > 5*s_)
for lab, kw in (("dt=0.05", {'dt': 0.05}), ("nu=0.5", {'nu': 0.5})):
    G_, c_ = run_case('A', 1.0, Ns, 600+len(lab), **kw); J_, s_ = current(G_, c_, Ns)
    check("(A), %s: J = 0 (|J|<4 sigma)" % lab, abs(J_) < 4*s_, "J=%+.4f±%.4f" % (J_, s_))
```

```sage
# --- 3f. Сравнение с «наивным» дрейфовым потоком автора: J_naive = n <v_d> h, v_d = kT B'/(|q| B^2) ---
def naive_flux(G):
    g, B0, ya, yb, yc = G['g'], G['B0'], G['ya'], G['yb'], G['yc']
    # (1/h) int_ya^yb  g B0 / (B0^2 (1+g(y-yc))^2) dy ;  умножаем на sqrt(2 pi) (нормировка на Gamma0 = n h/sqrt(2pi))
    integral = (1/(1+g*(ya-yc)) - 1/(1+g*(yb-yc)))/B0
    return math.sqrt(2*math.pi)*integral/(yb-ya)
for case in ('L', 'A', 'Babs'):
    G, cnt, J, s = RUNS[(case, 1.0)]
    print("%-5s B0=+1: наивный дрейфовый поток %+.3f Gamma0;  МК: %+.4f ± %.4f" % (case, naive_flux(G), J, s))
GL, _, JL, sL = RUNS[('L', 1.0)]
check("(L) идеализация автора: МК-ток < 2% от наивной оценки n<v_d>h", abs(JL) + 2*sL < 0.02*naive_flux(GL),
      "|J|+2sigma = %.4f vs %.3f" % (abs(JL)+2*sL, naive_flux(GL)))
```

### 3g. Что показало моделирование

* **(A)** (детальное равновесие на каждой стороне), **(B-refl)** (внешние стороны не эмитируют, но и не поглощают) и **(L)** (авторская идеализация: эмиссия только с внутренних сторон, пластины перекрывают всю область градиента) дают **J = 0 в пределах статистики** (σ ≈ 0.002–0.004 Γ₀, все |J| < 2.1σ) при B₀ = ±1 и 0. При этом наивный поток $n\langle v_d\rangle h$ равен 0.4–0.6 Γ₀. Дрейф ведущих центров в объёме гасится скачущими орбитами у стенок и обходом краёв, как требует $\mathbf j=\mathbf j_{gc}+c\nabla\times\mathbf M=0$. В (A) прямой перенос «внутр.1 → внутр.2» асимметричен по B, но эта асимметрия в точности компенсируется каналами через внешние стороны (правила сумм, взаимность Онзагера).
* **(B-abs)** (внешние стороны поглощают, но не эмитируют) даёт **ток, значимо отличный от нуля, нечётный по B** и меняющий знак вместе с B. Это ровно то, что «видит» автор. Но это граничное условие нарушает закон Кирхгофа (поглощательная способность ≠ эмиссионной), то есть внешние стороны работают как сток с μ → −∞. Термостат не гиббсов, условие Вегшайдера по циклу «эмиссия–перенос–поглощение» нарушено, плотность обеднена до ⟨n⟩ ≈ 0.39 от равновесной (п. 3d). Вечного двигателя здесь нет: ток задан неравновесным ГУ, введённым руками.
* Знак тока в (B-abs) в этой геометрии **противоположен** наивному дрейфу: электроны идут против **B×∇B**. При B₀ = ±2 ток практически исчезает (замагниченность). Величина и знак зависят от геометрии, поэтому они не являются универсальным предсказанием. Я проверил только одну геометрию.
* Самосогласованное поле я не моделировал. По разделу 2a гиббсово состояние стационарно при любом Φ(r) (пространственный заряд, контактные разности), поэтому в (A) учёт поля не может создать ток. Для (B-abs) поле изменит величину, но не устранит нарушение ГУ.

## 4. Оценка экспериментальных артефактов

**4a. Разность температур катодов.** Ричардсон–Дешман: $j=A T^2 e^{-\varphi/kT}$, $\frac{d\ln j}{dT}=\frac{2+\varphi/kT}{T}$. Площадь эмиттера: шайба диаметром 5 мм, S = 0.196 см². Материал по скану — LaB₆. Литературные параметры LaB₆ разбросаны: φ ≈ 2.6–2.9 эВ, A ≈ 29–82 А/(см²K²). Значение 29/2.66 (Lafferty, 1951) взято по памяти. Значения 41/2.61 (тонкие плёнки), 57/2.68 (110) и 82/2.86 (100) (Oshima и др., JAP 48, 3925, 1977) взяты из найденных аннотаций, сами статьи я не открывал. Для сравнения взят W: A = 60, φ = 4.5 эВ.

Чистый ток при разбалансе $I_{net}\approx\tau\,S\,[j(T+\Delta T)-j(T)]$, где τ — доля эмиссии, доходящая до другого катода (ограничена пространственным зарядом, неизвестна). Нижняя граница $\tau_{min}\approx I_{obs}/(jS)$ соответствует случаю, когда весь односторонний ток ≈ 2·10⁻⁸ А.

```sage
kB_eV = 8.617333e-5
S = math.pi*0.25**2            # см^2
T = 1200.0; I_obs = 2e-8; I_div = 1.5e-9
mats = [("LaB6 (Lafferty)", 29.0, 2.66), ("LaB6 плёнка", 41.0, 2.61), ("LaB6 (110)", 57.0, 2.68),
        ("LaB6 (100)", 82.0, 2.86), ("W", 60.0, 4.5)]
def jRD(A, phi, T): return A*T*T*math.exp(-phi/(kB_eV*T))
print("%-16s %10s %10s %9s %9s | ΔT [K] для I_net=2e-8 A при tau = 1, 1e-2, 1e-3, tau_min" % ("материал", "j, A/см2", "I_e=jS, A", "dlnj/dT", "j(1300)/j"))
DT = {}
for name, A, phi in mats:
    j = jRD(A, phi, T); Ie = j*S; s = (2+phi/(kB_eV*T))/T
    r100 = jRD(A, phi, T+100)/j
    row = []
    for tau in (1.0, 1e-2, 1e-3):
        x = I_obs/(tau*Ie)                               # нужный относительный разбаланс j
        row.append(math.log(1+x)/s if x < 1e6 else float('inf'))
    tau_min = min(1.0, I_obs/Ie)
    row.append(math.log(1+I_obs/(tau_min*Ie))/s)
    DT[name] = (j, Ie, s, row, tau_min)
    print("%-16s %10.3g %10.3g %9.4f %9.2f | %s  (tau_min=%.1g)" % (name, j, Ie, s, r100, "  ".join("%.3g" % v for v in row), tau_min))
for name in [m[0] for m in mats if m[0].startswith("LaB6")]:
    row = DT[name][3]
    check("%s: ΔT для 2e-8 A << 100 K даже при tau=1e-3 и tau=tau_min" % name, max(row[2], row[3]) < 100,
          "ΔT(tau=1)=%.2g K, ΔT(1e-3)=%.2g K, ΔT(tau_min)=%.2g K" % tuple(row[i] for i in (0, 2, 3)))
Ie_W = DT["W"][1]
check("W при 1200 K: полная эмиссия шайбы < 2e-8 A (эффект не мог бы питаться эмиссией W)", Ie_W < I_obs, "I_e(W) = %.2g A" % Ie_W)
# чувствительность нуля гальванометра (1 деление = 1.5e-9 A) в единицах ΔT, LaB6 (Lafferty), tau=1
j, Ie, s, row, tm = DT["LaB6 (Lafferty)"]
print("1 деление гальванометра (1.5e-9 A) <-> ΔT = %.2g K при tau=1;  %.2g K при tau=1e-2" % (I_div/(Ie*s), I_div/(1e-2*Ie*s)))
print("допуск ±100 K <-> отношение эмиссий j(1300)/j(1200) = %.1f, j(1100)/j(1200) = %.2f" % (jRD(29, 2.66, 1300)/j, jRD(29, 2.66, 1100)/j))
```

```sage
# --- 4b. Ноль при B=0 не означает равновесие: компенсация Δφ разностью ΔT ---
# одинаковые токи эмиссии при разных phi: A T1^2 exp(-phi1/kT1) = A T2^2 exp(-phi2/kT2)  ->  ΔT ≈ T Δφ/(φ + 2kT)
phi = 2.66
for dphi in (0.02, 0.05, 0.1):
    dT = T*dphi/(phi + 2*kB_eV*T)
    # точное решение методом бисекции
    lo, hi = 0.0, 500.0
    target = jRD(29, phi+dphi, T)
    for _ in range(80):
        mid = (lo+hi)/2
        if jRD(29, phi, T-mid) > target: lo = mid
        else: hi = mid
    print("Δφ = %.2f эВ  ->  ноль гальванометра при ΔT ≈ %.1f K (точно %.1f K)" % (dphi, dT, lo))
check("Δφ ~0.05-0.1 эВ (типичный разброс LaB6) компенсируется ΔT в пределах ±100 K", T*0.1/(phi+2*kB_eV*T) < 100)
print("Вывод 4b: после балансировки при B=0 катоды в общем случае НЕ в равновесии (T1≠T2, φ1≠φ2), а зеркальная")
print("симметрия диода нарушена. Тогда отклик на B, в том числе нечётный по B (тип термоэмиссионного")
print("преобразователя в магнитном поле), разрешён уже в первом порядке по ΔT. Нужный разбаланс ~1e-3..1 от I_e.")
```

```sage
# --- 4c. Эдс Нернста: V_N = N_coef * B * grad T * L ---
B_T = 0.01                     # Тл (на рис.8/табл.1 до ~70-100 Гс)
gradT = 1000.0/0.01            # K/м (1000 K на см)
L = 0.005                      # м (диаметр шайбы)
Rg = 1300.0                    # Ом (гальванометр)
j, Ie, s, row, tm = DT["LaB6 (Lafferty)"]
kT_e = kB_eV*T                 # В
Rd_min = kT_e/Ie               # дифф. сопротивление термоэмиссионного промежутка при обменном токе = полной эмиссии
Rd_sc = kT_e/I_obs             # если обменный ток ~ 2e-8 A (ограничение пространственным зарядом)
print("R_d = kT/(e I0): %.3g Ом (I0 = I_e = %.2g A) ... %.3g Ом (I0 = 2e-8 A)" % (Rd_min, Ie, Rd_sc))
print("напряжение, нужное для 2e-8 A: %.2g В (при R_d,min) ... %.2g В (при R_d,sc)" % (I_obs*(Rd_min+Rg), I_obs*(Rd_sc+Rg)))
Imax = 0.0
for Nc in (1e-8, 1e-7, 1e-6):
    VN = Nc*B_T*gradT*L
    I1 = VN/(Rd_min+Rg); I2 = VN/(Rd_sc+Rg); Imax = max(Imax, I1)
    print("N = %.0e В/(К·Тл): V_N = %.2g В -> I <= %.2g A (R_d,min) ... %.2g A (R_d,sc);  I/I_obs <= %.2g" % (Nc, VN, I1, I2, I1/I_obs))
check("эдс Нернста при N <= 1e-6 В/(К·Тл) даёт < 2e-8 A (не объясняет ток в одиночку)", Imax < I_obs, "I_max = %.2g A" % Imax)
print("Замечание: эдс Нернста нечётна по B и формально подходит по симметрии. При N ~ 1e-6 и L ~ 2 см (стаканы/вводы")
print("в поле магнита) и минимальном R_d верхняя оценка ~ %.0f%% от I_obs: полностью исключить вклад нельзя, но для объяснения" %
      (100*1e-6*B_T*gradT*0.02/(Rd_min+Rg)/I_obs))
print("всего тока нужны N >~ 3e-6 В/(К·Тл) и одновременно все остальные множители на верхнем пределе.")
```

## 5. Выводы

1. **Аналитика.** Максвелл–Больцман с однородной плотностью (и вообще $n_0e^{-q\Phi/kT}f_M$ при любом Φ) точно стационарен для уравнения Власова с любым B(r) и с термализующими столкновениями. Плотность тока в нём равна нулю в каждой точке, ∇B-дрейф (ф. 1 препринта) гасится током намагниченности. ГУ (A) и (B-refl) совместимы с этим состоянием. ГУ «внешние стороны поглощают, но не эмитируют» несовместимо: оно нарушает детальное равновесие (закон Кирхгофа для электронов).
2. **Моделирование.** (A), (B-refl) и (L) дают **J = 0** в пределах статистики (σ ≈ 0.003 одностороннего потока) при B = ±1 и 0. Наивный дрейфовый поток автора при этом 0.4–0.6. (B-abs) даёт ток ~0.09 Γ₀, нечётный по B. Это следствие неравновесного ГУ, а не механизм извлечения тепла при одной T. Знак тока в выбранной геометрии противоположен наивному дрейфу.
3. **Эксперимент.** Наблюдаемые 2·10⁻⁸ А составляют ~4·10⁻⁴ от полной эмиссии шайбы LaB₆ при 1200 K (~5·10⁻⁵ А). Такой ток соответствует ΔT ≈ 0.007–0.04 K при τ = 1 (разброс по параметрам LaB₆), ≈0.7–3.4 K при τ = 10⁻² и ≈30 K в самом пессимистичном случае τ = τ_min, когда весь односторонний ток ~2·10⁻⁸ А. Это во всех случаях меньше указанного допуска ±100 K. Ноль гальванометра при B = 0 не гарантирует равенства температур: разброс φ на 0.05–0.1 эВ компенсируется ΔT ≈ 20–40 K, и тогда нечётный по B отклик неравновесного диода разрешён. Эдс Нернста при заданных коэффициентах даёт ≲1.6·10⁻⁹ А для L = 5 мм (≲8% наблюдаемого). При L ~ 2 см верхняя оценка ~30%: исключить её нельзя, но как основное объяснение она маловероятна. W при 1200 K вообще не дал бы такого тока: вся эмиссия шайбы ~2·10⁻¹² А. **Вывод:** наблюдаемый ток по порядку величины легко объясняется неконтролируемой разностью температур / работ выхода в пределах допуска. Для его отличия от артефакта нужны независимые измерения T₁ − T₂ с точностью ≲0.1 K, перестановка катодов и обращение геометрии магнита. В препринте этого нет.

**Чего здесь нет / не проверено:** 3D-геометрия реальных шайб и стаканов; самосогласованное поле в МК; реальные значения τ, φ₁ − φ₂ и коэффициента Нернста для LaB₆; параметры LaB₆ из литературы не сверены с первоисточниками.
