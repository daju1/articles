# Казимир для тел, однородных вдоль оси: полная постановка (ТМ+ТЕ, проницаемые материалы)

**Зачем новый ноутбук.** В `Casimir_corner_Green_tensor_explained.ipynb` задача ставилась для идеального проводника и двух *независимых* скалярных полей (Дирихле для ТМ, Нейман для ТЕ), с тензором скалярного поля. Это частный случай. Здесь задача ставится заново, из уравнений Максвелла:

* тела — произвольные цилиндры (сечение — набор замкнутых контуров), из материала с $\varepsilon(\omega),\mu(\omega)$;
* обе поляризации решаются **совместно**. На границе проницаемого тела при $k_z\neq0$ они связаны, и связь учитывается точно;
* все заряды и векторные токи на поверхности восстанавливаются явно;
* идеальный проводник получается как предел $\varepsilon\to\infty$ (проверяется).

Тензор Максвелла, вакуумные средние и силы — следующие разделы (§6–§8), они будут добавлены отдельно. Здесь — постановка, интегральные уравнения и проверка решателя на точном решении.

**Соглашения** (те же, что в `rezonatory_i_volnovody/MenDrive_qnm_theory.ipynb`):

* ось $z$ — вдоль образующей тел (ось «волновода»); сечение — плоскость $(x,y)$;
* гауссова система, $\mathrm{rot}\,\mathbf E=\frac{i\omega}{c}\mu\mathbf H$, $\ \mathrm{rot}\,\mathbf H=-\frac{i\omega}{c}\varepsilon\mathbf E$;
* поля $\propto\mathbf E(x,y)\,e^{ik_zz}e^{-i\omega t}$; ниже $c=1$;
* $s$ — длина дуги вдоль контура сечения $\Gamma$; $\mathbf n$ — внешняя нормаль к телу; $\boldsymbol\tau=\hat{\mathbf z}\times\mathbf n$ — касательная (обход против часовой стрелки);
* $\partial_n=\mathbf n\cdot\nabla$, $\ \partial_s=\boldsymbol\tau\cdot\nabla$.

## 1. Поперечные поля через продольные

Из уравнений Максвелла в однородной области с $\varepsilon,\mu$ при зависимости $e^{ik_zz-i\omega t}$ все шесть компонент выражаются через $E_z$ и $H_z$. Обозначим
$$
\gamma^2=\varepsilon\mu\,\omega^2-k_z^2 .
$$
Тогда
$$
\mathbf E_\perp=\frac{i}{\gamma^2}\Big(k_z\nabla_\perp E_z-\omega\mu\,\hat{\mathbf z}\times\nabla_\perp H_z\Big),\qquad
\mathbf H_\perp=\frac{i}{\gamma^2}\Big(k_z\nabla_\perp H_z+\omega\varepsilon\,\hat{\mathbf z}\times\nabla_\perp E_z\Big),
$$
а сами $E_z,H_z$ удовлетворяют уравнению Гельмгольца $(\nabla_\perp^2+\gamma^2)\,u=0$.

**Вывод.** Покомпонентно (поперечные части роторов):
$$
\partial_yH_z-ik_zH_y=-i\omega\varepsilon E_x,\quad ik_zH_x-\partial_xH_z=-i\omega\varepsilon E_y,\quad
\partial_yE_z-ik_zE_y=i\omega\mu H_x,\quad ik_zE_x-\partial_xE_z=i\omega\mu H_y .
$$
Это четыре линейных уравнения на $E_x,E_y,H_x,H_y$; их решение — формулы выше. Продольные части роторов дают уравнение Гельмгольца. Ячейка ниже проверяет все шесть уравнений символьно (под Sage); я дополнительно проверил их численно на плоской волне $e^{i(px+qy)}$, $p^2+q^2=\gamma^2$: невязки $\sim10^{-16}$.

**Касательные компоненты на контуре.** Для границы с нормалью $\mathbf n$ и касательной $\boldsymbol\tau=\hat{\mathbf z}\times\mathbf n$ используем
$\boldsymbol\tau\cdot\nabla=\partial_s$, $\ \boldsymbol\tau\cdot(\hat{\mathbf z}\times\nabla)=\partial_n$:
$$
E_s=\boldsymbol\tau\cdot\mathbf E=\frac{i}{\gamma^2}\big(k_z\,\partial_sE_z-\omega\mu\,\partial_nH_z\big),\qquad
H_s=\boldsymbol\tau\cdot\mathbf H=\frac{i}{\gamma^2}\big(k_z\,\partial_sH_z+\omega\varepsilon\,\partial_nE_z\big).
$$

```sage
try:
    # символьная проверка §1 (только под Sage)
    var('x y kz w eps mu')
    Ez = function('Ez')(x, y); Hz = function('Hz')(x, y)
    g2 = eps*mu*w^2 - kz^2
    Ex = I/g2*(kz*diff(Ez, x) + w*mu*diff(Hz, y)); Ey = I/g2*(kz*diff(Ez, y) - w*mu*diff(Hz, x))
    Hx = I/g2*(kz*diff(Hz, x) - w*eps*diff(Ez, y)); Hy = I/g2*(kz*diff(Hz, y) + w*eps*diff(Ez, x))
    # rot H = -i w eps E  (x, y компоненты),  rot E = i w mu H
    r1 = (diff(Hz, y) - I*kz*Hy) + I*w*eps*Ex
    r2 = (I*kz*Hx - diff(Hz, x)) + I*w*eps*Ey
    r3 = (diff(Ez, y) - I*kz*Ey) - I*w*mu*Hx
    r4 = (I*kz*Ex - diff(Ez, x)) - I*w*mu*Hy
    # z-компоненты: rot_z H = -i w eps E_z, rot_z E = i w mu H_z  -> уравнения Гельмгольца с gamma^2
    r5 = (diff(Hy, x) - diff(Hx, y)) + I*w*eps*Ez
    r6 = (diff(Ey, x) - diff(Ex, y)) - I*w*mu*Hz
    print('поперечные (r1..r4):', [bool(r.full_simplify() == 0) for r in (r1, r2, r3, r4)])
    print('r5 = 0  <=>  lap Ez + g2 Ez = 0:', bool((r5*g2/(I*w*eps) - (diff(Ez, x, 2) + diff(Ez, y, 2) + g2*Ez)).full_simplify() == 0))
    print('r6 = 0  <=>  lap Hz + g2 Hz = 0:', bool((-r6*g2/(I*w*mu) - (diff(Hz, x, 2) + diff(Hz, y, 2) + g2*Hz)).full_simplify() == 0))
except NameError:
    print('символьная проверка требует Sage; численная проверка — ниже, §5')
```

## 2. Граничные условия и почему поляризации связаны

На границе двух сред непрерывны касательные компоненты: $E_z,\ H_z,\ E_s,\ H_s$. С формулами §1 условия для $E_s,H_s$ дают
$$
\frac{1}{\gamma_1^2}\big(k_z\partial_sE_z-\omega\mu_1\partial_nH_z\big)\Big|_1=\frac{1}{\gamma_2^2}\big(k_z\partial_sE_z-\omega\mu_2\partial_nH_z\big)\Big|_2,
$$
$$
\frac{1}{\gamma_1^2}\big(k_z\partial_sH_z+\omega\varepsilon_1\partial_nE_z\big)\Big|_1=\frac{1}{\gamma_2^2}\big(k_z\partial_sH_z+\omega\varepsilon_2\partial_nE_z\big)\Big|_2 .
$$
$E_z$ и $H_z$ непрерывны, поэтому $\partial_s$ одинаковы с обеих сторон. Если $\gamma_1\neq\gamma_2$ и $k_z\neq0$, член $k_z\partial_sE_z(1/\gamma_1^2-1/\gamma_2^2)$ входит в условие на $\partial_nH_z$, и наоборот. **ТМ и ТЕ связаны.** Развязка бывает только в двух случаях:

* $k_z=0$ (волна строго поперёк оси);
* идеальный проводник: $E_z=0$, значит $\partial_sE_z=0$, и остаётся $\partial_nH_z=0$ — точная развязка при любой форме сечения (это использовалось в старом ноутбуке).

В вакуумном среднем интегрируют по всем $k_z$, поэтому для реальных материалов связь существенна.

## 3. Поверхностные заряды и векторные токи

Для идеального проводника (а для проницаемого тела — как скачки полей) поверхностный заряд и ток
$$
\eta=\frac{1}{4\pi}\,[\mathbf n\cdot\mathbf D],\qquad \mathbf K=\frac{c}{4\pi}\,\mathbf n\times[\mathbf H],
$$
где $[\cdot]$ — скачок «снаружи минус внутри». У проницаемого тела $[\mathbf n\cdot\mathbf D]=0$ и $[\mathbf H_t]=0$: свободных поверхностных зарядов и токов нет, есть объёмные поляризационные токи. У идеального проводника внутри поля нет, и через поля снаружи ($\gamma=\gamma_0$, вакуум):

| | заряд $\eta$ | ток $K_s$ (по контуру) | ток $K_z$ (вдоль оси) |
|---|---|---|---|
| ТМ ($H_z=0$) | $\dfrac{i k_z}{4\pi\gamma_0^2}\,\partial_nE_z$ | $0$ | $\dfrac{i\omega}{4\pi\gamma_0^2}\,\partial_nE_z$ |
| ТЕ ($E_z=0$) | $-\dfrac{i\omega}{4\pi\gamma_0^2}\,\partial_sH_z$ | $-\dfrac{1}{4\pi}H_z$ | $\dfrac{i k_z}{4\pi\gamma_0^2}\,\partial_sH_z$ |

(знаки — для $\mathbf K=\mathbf n\times\mathbf H/4\pi$ с $\mathbf n\times\hat{\mathbf z}=-\boldsymbol\tau$, $\mathbf n\times\boldsymbol\tau=\hat{\mathbf z}$).
Уравнение непрерывности $-i\omega\eta+\partial_sK_s+ik_zK_z=0$ выполняется тождественно в обоих случаях — проверка ниже.

```sage
try:
    var('kz w g0 s')
    q = function('q')(s); Hs_ = function('Hz')(s)
    # ТМ: q = dE_z/dn
    eta = I*kz/(4*pi*g0^2)*q; Ks = 0; Kz = I*w/(4*pi*g0^2)*q
    print('ТМ: непрерывность =', bool((-I*w*eta + diff(Ks, s) + I*kz*Kz).full_simplify() == 0))
    # ТЕ: g0^2 = w^2 - kz^2 в вакууме
    eta = -I*w/(4*pi*g0^2)*diff(Hs_, s); Ks = -Hs_/(4*pi); Kz = I*kz/(4*pi*g0^2)*diff(Hs_, s)
    expr = (-I*w*eta + diff(Ks, s) + I*kz*Kz).subs(g0^2 == w^2 - kz^2)
    print('ТЕ: непрерывность =', bool(expr.full_simplify() == 0))
except NameError:
    print('требует Sage')
```

## 4. Переход на мнимую частоту и интегральные уравнения

Вакуумные средние считаются на мнимой частоте $\omega=i\xi$, $\xi>0$. Тогда
$$
\gamma^2=\varepsilon\mu\,\omega^2-k_z^2=-\kappa^2,\qquad \kappa^2=\varepsilon(i\xi)\mu(i\xi)\,\xi^2+k_z^2 ,
$$
и в каждой области $(-\nabla_\perp^2+\kappa^2)u=0$. Для всех причинных материалов $\varepsilon(i\xi),\mu(i\xi)$ вещественны и $\geq1$. Модели (класс `Material`):

* вакуум: $\varepsilon=\mu=1$;
* диэлектрик: $\varepsilon=\mathrm{const}$;
* плазма: $\varepsilon(i\xi)=1+\omega_p^2/\xi^2$;
* Друде: $\varepsilon(i\xi)=1+\omega_p^2/\big(\xi(\xi+\gamma)\big)$;
* осциллятор Лоренца: $\varepsilon(i\xi)=1+(\varepsilon_0-1)\,\omega_0^2/(\omega_0^2+\xi^2)$.

**Удобная переменная.** При $\omega=i\xi$ множители $i$ в формулах §1 становятся вещественными, если взять $H_z=-i\,h$. Тогда условия §2 (обозначаем $q=\partial_n$ снаружи, $\partial_n^{\,\rm in}$ — изнутри):
$$
\partial_n^{\rm in}E_z=\frac{\varepsilon_o\kappa_i^2}{\varepsilon_i\kappa_o^2}\,\partial_nE_z+\frac{k_z}{\varepsilon_i\,\xi}\Big(1-\frac{\kappa_i^2}{\kappa_o^2}\Big)\partial_s h,\qquad
\partial_n^{\rm in}h=\frac{\mu_o\kappa_i^2}{\mu_i\kappa_o^2}\,\partial_nh+\frac{k_z}{\mu_i\,\xi}\Big(1-\frac{\kappa_i^2}{\kappa_o^2}\Big)\partial_s E_z .
$$
(индексы $o$ — снаружи, $i$ — внутри тела). Все коэффициенты вещественны.

**Формула Грина.** Функция Грина $G_\kappa(r)=K_0(\kappa r)/2\pi$ (вывод — §3 старого ноутбука). Для решения $u$ уравнения Гельмгольца снаружи (затухающего на бесконечности) и внутри:
$$
\tfrac12u=-\hat S_{o}\,\partial_nu+\hat D_{o}\,u+u^{\rm inc}\quad\text{(снаружи)},\qquad
\tfrac12u=\hat S_{i}\,\partial_n^{\rm in}u-\hat D_{i}\,u\quad\text{(изнутри)},
$$
где $(\hat Sf)(x)=\int_\Gamma G(x-y)f(y)\,ds_y$, $\ (\hat Df)(x)=\mathrm{v.p.}\!\int_\Gamma\partial_{n_y}G(x-y)\,f(y)\,ds_y$. Подставляя связь $\partial_n^{\rm in}$ с внешними величинами, получаем систему $4N\times4N$ на неизвестные $E_z,\,h,\,\partial_nE_z,\,\partial_nh$ на контуре. Правая часть — падающее поле источника.

**Дискретизация** — те же панели (§5 старого ноутбука): значения постоянны на панели, коллокация в серединах; $\partial_s$ — разностная производная по соседним серединам вдоль замкнутого контура.

```sage
import numpy as np, time
pi = np.pi
def K0K1(x):
    """K0(x), K1(x) для x>0 — Абрамовиц–Стиган 9.8.1-9.8.8."""
    x = np.asarray(x, float); K0 = np.empty_like(x); K1 = np.empty_like(x)
    s = x <= 2.0; t = x[s]/2; t2 = (x[s]/3.75)**2
    I0 = 1 + t2*(3.5156229 + t2*(3.0899424 + t2*(1.2067492 + t2*(0.2659732 + t2*(0.0360768 + t2*0.0045813)))))
    I1 = x[s]*(0.5 + t2*(0.87890594 + t2*(0.51498869 + t2*(0.15084934 + t2*(0.02658733 + t2*(0.00301532 + t2*0.00032411))))))
    y = t*t
    K0[s] = -np.log(t)*I0 + (-0.57721566 + y*(0.42278420 + y*(0.23069756 + y*(0.03488590 + y*(0.00262698 + y*(0.00010750 + y*0.0000074))))))
    K1[s] = np.log(t)*I1 + (1/x[s])*(1 + y*(0.15443144 + y*(-0.67278579 + y*(-0.18156897 + y*(-0.01919402 + y*(-0.00110404 + y*(-0.00004686)))))))
    b = ~s; xb = x[b]; z = 2/xb; e = np.exp(-xb)/np.sqrt(xb)
    K0[b] = e*(1.25331414 + z*(-0.07832358 + z*(0.02189568 + z*(-0.01062446 + z*(0.00587872 + z*(-0.00251540 + z*0.00053208))))))
    K1[b] = e*(1.25331414 + z*(0.23498619 + z*(-0.03655620 + z*(0.01504268 + z*(-0.00780353 + z*(0.00325614 + z*(-0.00068245)))))))
    return K0, K1

def k0_selfpanel(kap, h):
    """S_jj: (1/2pi) * интеграл K0(kap|t|) по панели длины h (особенность ln вычтена аналитически)."""
    xg, wg = np.polynomial.legendre.leggauss(16)
    T = h/2; t = (xg+1)/2*T; w = wg/2*T
    K0, _ = K0K1(kap*t)
    reg = K0 + np.log(kap*t/2)
    return (np.sum(w*reg) - (T*np.log(kap*T/2) - T)) / pi

def panels(P0, P1, N):
    """N панелей на отрезке P0-P1, сгущение по косинусу к концам. Возвращает (середины, длины)."""
    u = (1 - np.cos(np.linspace(0, pi, N+1))) / 2
    P0 = np.asarray(P0, float); P1 = np.asarray(P1, float)
    nodes = P0[None, :] + u[:, None]*(P1 - P0)[None, :]
    return (nodes[1:] + nodes[:-1])/2, np.linalg.norm(nodes[1:] - nodes[:-1], axis=1)

def Smat(mid, h, kap):
    """Матрица простого слоя S_jb."""
    R = np.linalg.norm(mid[:, None, :] - mid[None, :, :], axis=2); np.fill_diagonal(R, 1.0)
    K0, _ = K0K1(kap*R); S = K0/(2*pi)*h[None, :]
    for j in range(len(h)): S[j, j] = k0_selfpanel(kap, h[j])
    return S

def G0grad(r, s, k):
    """g0(r_m - s_b) и его градиент по r: массивы (M,B) и (M,B,2)."""
    d = r[:, None, :] - s[None, :, :]; R = np.linalg.norm(d, axis=2)
    K0, K1 = K0K1(k*R)
    return K0/(2*pi), (-k*K1/(2*pi*R))[..., None]*d

def poly_body(V, Ns):
    """Многоугольник с вершинами V (против часовой), Ns панелей на сторонах: (середины, длины, внешние нормали)."""
    M = []; H = []; NN = []
    for i in range(len(V)):
        X = np.asarray(V[i], float); Y = np.asarray(V[(i+1) % len(V)], float)
        m, h = panels(X, Y, Ns[i]); tau = (Y-X)/np.linalg.norm(Y-X)
        M.append(m); H.append(h); NN.append(np.tile([tau[1], -tau[0]], (Ns[i], 1)))
    return np.vstack(M), np.concatenate(H), np.vstack(NN)

def kgrid(n=60, s=1.0):
    """Узлы и веса для интеграла по kappa от 0 до бесконечности: kappa = s*tan(pi u/2)."""
    u = (np.arange(n) + 0.5)/n
    return s*np.tan(pi/2*u), s*pi/2/np.cos(pi/2*u)**2/n
```

```sage
# ---------------- материалы на мнимой частоте xi (eps, mu вещественны и >= 1) ----------------
class Material:
    """eps(xi), mu(xi) на мнимой частоте omega = i*xi. kind='pec' — идеальный проводник."""
    def __init__(self, kind='vacuum', eps=1.0, mu=1.0, wp=None, gamma=0.0, w0=None):
        self.kind, self.e0, self.m0, self.wp, self.g, self.w0 = kind, eps, mu, wp, gamma, w0
    def eps(self, xi):
        if self.kind in ('vacuum',): return 1.0
        if self.kind == 'dielectric': return self.e0
        if self.kind == 'plasma': return 1.0 + self.wp**2/xi**2
        if self.kind == 'drude':  return 1.0 + self.wp**2/(xi*(xi + self.g))
        if self.kind == 'lorentz': return 1.0 + (self.e0 - 1.0)*self.w0**2/(self.w0**2 + xi**2)
        raise ValueError(self.kind)
    def mu(self, xi): return 1.0 if self.kind != 'dielectric' else self.m0

def Kmat(mid, h, nrm, k):
    """Двойной слой: K_jb = h_b * d/dn_b G(m_j - m_b),  K_jj = 0 (прямая панель)."""
    d = mid[:, None, :] - mid[None, :, :]; R = np.linalg.norm(d, axis=2); np.fill_diagonal(R, 1.0)
    _, K1 = K0K1(k*R); gr = (-k*K1/(2*pi*R))[..., None]*d          # grad_x G(x - y)
    Km = -np.einsum('jbc,bc->jb', gr, nrm)*h[None, :]; np.fill_diagonal(Km, 0.0)
    return Km

def Dtau(mid, h, nrm, loops):
    """Производная по дуге d/ds (s вдоль tau = z x n) на замкнутых контурах: центральные разности
    по серединам панелей. loops — список массивов индексов панелей каждого замкнутого контура (по порядку)."""
    N = len(h); D = np.zeros((N, N))
    for L in loops:
        n = len(L)
        for a in range(n):
            j, jp, jm = L[a], L[(a+1) % n], L[(a-1) % n]
            sp = (h[j] + h[jp])/2; sm = (h[j] + h[jm])/2
            # трёхточечная формула на неравномерной сетке
            D[j, jp] += sm/(sp*(sp+sm)); D[j, jm] += -sp/(sm*(sp+sm)); D[j, j] += (sp - sm)/(sp*sm)
    return D

def solve_penetrable(body, loops, mat_in, mat_out, xi, kz, Einc, hinc):
    """Связанная задача ТМ+ТЕ для z-однородного тела из материала mat_in в среде mat_out.
    Неизвестные на контуре: E_z, h (H_z = -i h), dE_z/dn, dh/dn снаружи.  Возвращает (E, h, qE, qh)."""
    mid, h, nrm = body[:3]; N = len(h)
    eo, mo, ei, mi = mat_out.eps(xi), mat_out.mu(xi), mat_in.eps(xi), mat_in.mu(xi)
    ko = np.sqrt(eo*mo*xi**2 + kz**2); ki = np.sqrt(ei*mi*xi**2 + kz**2)
    So, Ko = Smat(mid, h, ko), Kmat(mid, h, nrm, ko); Si, Ki = Smat(mid, h, ki), Kmat(mid, h, nrm, ki)
    Dt = Dtau(mid, h, nrm, loops); I = np.eye(N); Z = np.zeros((N, N))
    r2 = ki**2/ko**2; c = (kz/xi)*(1 - r2)
    # q_Ei = aE qEo + bE Dt h ;  q_hi = ah qho + bh Dt E
    aE, bE = eo*r2/ei, c/ei
    ah, bh = mo*r2/mi, c/mi
    # строки: [ext E], [ext h], [int E], [int h];  столбцы: [E, h, qEo, qho]
    A = np.block([[0.5*I - Ko, Z, So, Z],
                  [Z, 0.5*I - Ko, Z, So],
                  [0.5*I + Ki, -Si @ (bE*Dt), -Si*aE, Z],
                  [-Si @ (bh*Dt), 0.5*I + Ki, Z, -Si*ah]])
    rhs = np.concatenate([Einc, hinc, np.zeros(N), np.zeros(N)])
    x = np.linalg.solve(A, rhs)
    return x[:N], x[N:2*N], x[2*N:3*N], x[3*N:], (ko, ki)

def field_out(body, sol, pts):
    """Рассеянные E_z, h во внешних точках по формуле Грина: u = -S q + D u (знак для n наружу)."""
    mid, h, nrm = body[:3]; E, hh, qE, qh, (ko, ki) = sol
    Gm, gr = G0grad(pts, mid, ko)
    dnG = -np.einsum('mbc,bc->mb', gr, nrm)          # d/dn_y G(x-y)
    Es = (dnG*h[None, :]) @ E - (Gm*h[None, :]) @ qE
    hs = (dnG*h[None, :]) @ hh - (Gm*h[None, :]) @ qh
    return Es, hs
```

## 5. Проверка решателя на точном решении: круглый цилиндр

Для круга радиуса $a$ решение раскладывается по гармоникам $e^{in\varphi}$: внутри — $I_n(\kappa_ir)$, снаружи — $K_n(\kappa_or)$. Падающее поле — точечный источник $E_z$ на расстоянии $r_s$:
$G(\mathbf x-\mathbf r_s)=\frac1{2\pi}\sum_n I_n(\kappa_or)K_n(\kappa_or_s)e^{in\varphi}$ при $r<r_s$. Для каждого $n$ — система $4\times4$ из непрерывности $E_z,h$ и условий §4 (в форме §2). Она точно учитывает связь ТМ и ТЕ. Сравниваем рассеянные $E_z$ и $h$ в трёх точках при $r=0{,}9$.

```sage
def bessel_IK(nmax, x):
    """I_n(x), K_n(x), n=0..nmax, x>0 скаляр. K — вверх по рекуррентности, I — вниз (Миллер) с нормировкой I0."""
    K0, K1 = K0K1(np.array([x])); K = np.zeros(nmax+2); K[0], K[1] = K0[0], K1[0]
    for n in range(1, nmax+1): K[n+1] = K[n-1] + 2*n/x*K[n]
    M = nmax + 40 + int(2*x); Iv = np.zeros(M+2); Iv[M] = 1e-300
    for n in range(M, 0, -1): Iv[n-1] = Iv[n+1] + 2*n/x*Iv[n]
    t = 1.0; m = 0; I0 = 0.0
    while True:                                   # I0 рядом
        I0 += t; m += 1; t *= (x/2)**2/m**2
        if t < 1e-17*I0: break
    I = Iv[:nmax+2]*(I0/Iv[0])
    return I[:nmax+1], K[:nmax+1]

def exact_circle(a, mat_in, mat_out, xi, kz, rs, pts, nmax=40):
    """Точное решение: точечный источник E_z = G_o(x - rs) (rs на оси x, rs>a), круглый цилиндр радиуса a.
    Возвращает рассеянные E_z, h в точках pts (вне цилиндра, |x|<rs не требуется)."""
    eo, mo, ei, mi = mat_out.eps(xi), mat_out.mu(xi), mat_in.eps(xi), mat_in.mu(xi)
    ko = np.sqrt(eo*mo*xi**2 + kz**2); ki = np.sqrt(ei*mi*xi**2 + kz**2)
    Io, Ko_ = bessel_IK(nmax+1, ko*a); Ii, Ki_ = bessel_IK(nmax+1, ki*a); _, Ks = bessel_IK(nmax+1, ko*rs)
    rr = np.linalg.norm(pts, axis=1); ph = np.arctan2(pts[:, 1], pts[:, 0])
    Es = np.zeros(len(pts), complex); hs = np.zeros(len(pts), complex)
    _, Kr = zip(*[bessel_IK(nmax+1, ko*r) for r in rr]); Kr = np.array(Kr)
    xo, xi_ = ko*a, ki*a
    for n in range(-nmax, nmax+1):
        m = abs(n)
        In_o, Kn_o, In_i = Io[m], Ko_[m], Ii[m]
        dIn_o = ko*(Io[m+1] + m/xo*Io[m]); dKn_o = ko*(-Ko_[m+1] + m/xo*Ko_[m]); dIn_i = ki*(Ii[m+1] + m/xi_*Ii[m])
        cin = Ks[m]/(2*pi)                      # коэффициент падающей при I_n(ko r) e^{in phi}
        dt = 1j*n/a                             # d/ds = (1/a) d/dphi
        # неизвестные: aE (вне K_n), bh (вне K_n), cE (вн I_n), dh (вн I_n)
        # E: cin In_o + aE Kn_o = cE In_i ;  h: bh Kn_o = dh In_i
        # (1/ko^2)(kz dt E - xi mo dh/dr)_out = (1/ki^2)(kz dt E - xi mi dh/dr)_in
        # (1/ko^2)(kz dt h - xi eo dE/dr)_out = (1/ki^2)(kz dt h - xi ei dE/dr)_in
        A = np.array([[Kn_o, 0, -In_i, 0],
                      [0, Kn_o, 0, -In_i],
                      [kz*dt*Kn_o/ko**2, -xi*mo*dKn_o/ko**2, -kz*dt*In_i/ki**2, xi*mi*dIn_i/ki**2],
                      [-xi*eo*dKn_o/ko**2, kz*dt*Kn_o/ko**2, xi*ei*dIn_i/ki**2, -kz*dt*In_i/ki**2]], complex)
        b = -cin*np.array([In_o, 0, kz*dt*In_o/ko**2, -xi*eo*dIn_o/ko**2], complex)
        aE, bh, cE, dh = np.linalg.solve(A, b)
        Es += aE*Kr[:, m]*np.exp(1j*n*ph); hs += bh*Kr[:, m]*np.exp(1j*n*ph)
    return Es, hs
```

```sage
a = 0.5; rs = np.array([1.2, 0.0])
pts = np.array([[0.9*np.cos(t), 0.9*np.sin(t)] for t in (0.4, 1.3, 2.5)])
mo = Material()
cases = [('диэлектрик eps=4', Material('dielectric', eps=4.0)),
         ('Друде wp=5, gamma=0.3', Material('drude', wp=5.0, gamma=0.3)),
         ('плазма wp=60 (почти идеальный)', Material('plasma', wp=60.0))]
xi, kz = 1.0, 0.8
for Np in (100, 200, 400, 800):
    th = np.linspace(0, 2*pi, Np+1); V = np.stack([np.cos(th[:-1]), np.sin(th[:-1])], 1)*a
    body = poly_body(V, [1]*Np); loops = [np.arange(Np)]
    ko = np.sqrt(xi**2 + kz**2); Einc = G0grad(body[0], rs[None, :], ko)[0][:, 0]
    row = []
    for name, mat in cases:
        sol = solve_penetrable(body, loops, mat, mo, xi, kz, Einc, np.zeros(Np))
        Es, hs = field_out(body, sol, pts); Ex, hx = exact_circle(a, mat, mo, xi, kz, rs[0], pts)
        row.append('%s: dE %.1e, dh %.1e' % (name.split()[0], np.max(abs(Es - Ex.real))/np.max(abs(Ex)), np.max(abs(hs - hx.real))/np.max(abs(hx))))
    print('N=%4d   ' % Np + ' | '.join(row))
```

* Точное решение для круга строится из тех же условий непрерывности (§2, §4), что и интегральное уравнение. Поэтому сравнение проверяет реализацию интегральных уравнений и дискретизацию, а сами условия §2 опираются на формулы §1 (проверены отдельно).
* Решатель сходится к точному решению как $1/N$ для всех трёх материалов, **включая связанную ТЕ-компоненту** $h$. Её создаёт чисто ТМ-источник только из-за связи на границе при $k_z\neq0$. Это прямая численная проверка §2.
* Порядок $1/N$ — следствие кусочно-постоянных значений на панелях. Его можно повысить, но для постановки задачи этого достаточно.
* Для плазмы с $\omega_p=60$ (скин-слой $1/\omega_p\approx0{,}017\ll a$) $h$ сходится медленнее: поле внутри сосредоточено в тонком слое у поверхности, который панели должны разрешать. Предел идеального проводника для таких тел удобнее брать отдельной формулировкой (условия §2 при $E_z=0$), как в старом ноутбуке.
