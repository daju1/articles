# Сила Казимира на уголок: функции Грина и тензор натяжений от А до Я

**Что в этом ноутбуке.** Подробно, с выводами и проверками, воспроизводится расчёт полной вакуумной силы на уголок из двух крыльев конечной толщины. Полная сила считается двумя способами через поток тензора натяжений $\langle T_{ik}\rangle$:

* **Вариант 1:** интеграл тензора по большой окружности, охватывающей весь уголок.
* **Вариант 2:** интеграл тензора по контуру, обжатому вплотную вокруг каждого крыла (на расстоянии $\varepsilon$ от металла), с разбивкой на грани и торцы.

Оба варианта сверяются с независимым **энергетическим методом** (формула Ллойда / TGTG, $F=-\partial E/\partial X$).

**Модель (ограничения):** геометрия однородна вдоль $z$ (2D-сечение), ТМ-поляризация ↔ скалярное поле с условием Дирихле, идеальный проводник, $T=0$, статика. Единицы $\hbar=c=1$; сила — на единицу длины вдоль $z$.

План:
1. Почему ТМ-задача сводится к скалярному полю Дирихле
2. Евклидов поворот: сумма по модам → интеграл по $\kappa$
3. Свободная функция Грина $K_0(\kappa r)/2\pi$ (вывод)
4. Функция Грина при наличии проводников: простой слой
5. Дискретизация: панели, регуляризация самодействия
6. Тензор натяжений через функцию Грина; сохранение $\partial_k T_{ik}=0$
7. Сила как поток тензора; почему контур можно деформировать
8. Энергетический метод (Ллойд/TGTG) — независимая проверка
9. Тесты: функция $K_0$, условие Дирихле, $\partial_kT_{ik}=0$, два тела
10. Уголок: вариант 1 (окружность), вариант 2 (поверхность крыльев)
11. Итоги и что из них следует

## 1. Сведение к скалярному полю

Пусть проводники — цилиндры произвольного сечения, однородные вдоль $z$. Тогда поле раскладывается на две независимые поляризации:

* **ТМ** ($H_z=0$): всё определяется $E_z=\varphi(x,y)\,e^{ik_zz-i\omega t}$. На идеальном проводнике касательное $E$ равно нулю, значит $\varphi|_{\partial}=0$ — **условие Дирихле**.
* **ТЕ** ($E_z=0$): всё определяется $H_z$, условие Неймана $\partial_n H_z|_\partial=0$.

Каждая компонента удовлетворяет $(\nabla_\perp^2 - k_z^2 + \omega^2)\varphi=0$. Значит, ТМ-вклад в энергию и силу совпадает с вкладом **безмассового скалярного поля в 3D** с условием Дирихле на поверхности цилиндров. Это и считается ниже; ТЕ-вклад (Нейман) сюда не входит.

Плотность лагранжиана скаляра: $\mathcal L=\tfrac12(\dot\varphi^2-(\nabla\varphi)^2)$. Канонический тензор энергии-импульса (пространственная часть):
$$
\Theta_{ij}=\partial_i\varphi\,\partial_j\varphi-\tfrac12\delta_{ij}\big((\nabla\varphi)^2-\dot\varphi^2\big),
\qquad \partial_t T^{0i}+\partial_j\Theta_{ij}=0 .
$$
Тензор натяжений (максвелловский знак) — $\sigma_{ij}=-\Theta_{ij}$. Сила на тело внутри замкнутой поверхности: $F_i=\oint \sigma_{ij}n_j\,dS$.

## 2. Евклидов поворот: от суммы по модам к интегралу по $\kappa$

Вакуумное среднее $\langle\varphi(x)\varphi(x')\rangle$ выражается через функцию Грина. После поворота Вика $\omega=i\xi$ уравнение для фурье-компоненты по $t$ и $z$ принимает вид
$$
\big(-\nabla_\perp^2+\xi^2+k_z^2\big)\,g(\mathbf r,\mathbf r')=\delta^2(\mathbf r-\mathbf r'),\qquad g|_{\partial}=0 .
$$
Оператор зависит только от $\kappa^2=\xi^2+k_z^2$. Поэтому
$$
\langle\varphi(x)\varphi(x')\rangle_E=\int\frac{d\xi\,dk_z}{(2\pi)^2}\,e^{i\xi(\tau-\tau')+ik_z(z-z')}\,g_\kappa(\mathbf r,\mathbf r'),
\qquad \int\frac{d\xi\,dk_z}{(2\pi)^2}\,f(\kappa)=\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}f(\kappa).
$$
Средние билинейных величин в совпадающих точках:

| величина | вклад под интегралом |
|---|---|
| $\langle\partial_i\varphi\,\partial_j\varphi\rangle$ | $\partial_i\partial'_j g_\kappa$ |
| $\langle(\partial_z\varphi)^2\rangle$ | $k_z^2\,g_\kappa$ |
| $\langle\dot\varphi^2\rangle$ | $-\xi^2\,g_\kappa$ (знак от поворота $\partial_t\to i\partial_\tau$) |

В $\Theta_{ij}$ (для $i,j\in\{x,y\}$) под $\delta_{ij}$ стоит $(\nabla\varphi)^2-\dot\varphi^2=(\nabla_\perp\varphi)^2+(\partial_z\varphi)^2-\dot\varphi^2\to \partial_k\partial'_k g+(k_z^2+\xi^2)g$. Отсюда
$$
\boxed{\;\langle\Theta_{ij}(\mathbf r)\rangle=\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}\Big[D_{ij}-\tfrac12\delta_{ij}\big(D_{kk}+\kappa^2 g_s\big)\Big],\quad D_{ij}=\partial_i\partial'_j g_s(\mathbf r,\mathbf r')\big|_{\mathbf r'=\mathbf r}\;}
$$
Здесь $g_s=g-g_0$ — **рассеянная часть**. Вычитание свободной $g_0$ убирает бесконечную, но однородную плотность «пустого» вакуума: её дивергенция равна нулю, и в силу она не входит. Остаток конечен везде вне проводников.

## 3. Свободная функция Грина: $g_0=K_0(\kappa r)/2\pi$

Решаем $(-\nabla^2+\kappa^2)g_0=\delta^2(\mathbf r)$ на плоскости. Фурье: $g_0=\int\frac{d^2q}{(2\pi)^2}\frac{e^{i\mathbf q\mathbf r}}{q^2+\kappa^2}$. Угловой интеграл даёт $J_0(qr)$:
$$
g_0=\frac1{2\pi}\int_0^\infty\frac{q\,J_0(qr)}{q^2+\kappa^2}\,dq=\frac{1}{2\pi}K_0(\kappa r).
$$
(Табличный интеграл; проверить можно иначе: при $r\neq0$ функция $K_0(\kappa r)$ решает модифицированное уравнение Бесселя $g''+g'/r-\kappa^2g=0$ и убывает на бесконечности, а при $r\to0$ $K_0\approx-\ln(\kappa r/2)-\gamma$. Поток $-\oint\partial_r g_0\,dl=-2\pi r\cdot(-1/2\pi r)=1$ даёт правильную нормировку дельта-функции.)

Нужные формулы:
* $K_0'(x)=-K_1(x)$, поэтому $\nabla_{\mathbf r}\,g_0(\mathbf r-\mathbf s)=-\dfrac{\kappa K_1(\kappa R)}{2\pi R}(\mathbf r-\mathbf s)$, где $R=|\mathbf r-\mathbf s|$.
* Интегральное представление (для проверки): $K_0(x)=\int_0^\infty e^{-x\cosh t}dt$, $K_1(x)=\int_0^\infty\cosh t\,e^{-x\cosh t}dt$.

В коде $K_0,K_1$ вычисляются полиномиальными аппроксимациями Абрамовица–Стиган (9.8.1–9.8.8), точность ~$10^{-7}$. Scipy не нужен.

```sage
try:
    preparser(False)
except NameError:
    pass
import numpy as np, time
pi = np.pi
trapz = getattr(np, 'trapezoid', None) or np.trapz
def check(name, residual, tol):
    r = float(abs(residual))
    print(('OK  ' if r < tol else 'FAIL') + '  %-66s residual = %.3e (tol %.0e)' % (name, r, tol))
```

```sage
def K0K1(x):
    """K0(x), K1(x) для x>0 — Абрамовиц–Стиган 9.8.1-9.8.8."""
    x = np.asarray(x, float); K0 = np.empty_like(x); K1 = np.empty_like(x)
    s = x <= 2.0; t = x[s]/2; t2 = (x[s]/3.75)**2
    I0 = 1 + t2*(3.5156229 + t2*(3.0899424 + t2*(1.2067492 + t2*(0.2659732 + t2*(0.0360768 + t2*0.0045813)))))
    I1 = x[s]*(0.5 + t2*(0.87890594 + t2*(0.51498869 + t2*(0.15084934 + t2*(0.02658733 + t2*(0.00301532 + t2*0.00032411))))))
    y = t*t
    K0[s] = -np.log(t)*I0 + (-0.57721566 + y*(0.42278420 + y*(0.23069756 + y*(0.03488590 + y*(0.00262698 + y*(0.00010750 + y*0.0000074))))))
    K1[s] = np.log(t)*I1 + (1/x[s])*(1 + y*(0.15443144 + y*(-0.67278579 + y*(-0.18156897 + y*(-0.01919402 + y*(-0.00110404 + y*(-0.00004686)))))))
    b = ~s; xb = x[b]; z = 2/xb; e = np.exp(-xb)/np.sqrt(xb)
    K0[b] = e*(1.25331414 + z*(-0.07832358 + z*(0.02189568 + z*(-0.01062446 + z*(0.00587872 + z*(-0.00251540 + z*0.00053208))))))
    K1[b] = e*(1.25331414 + z*(0.23498619 + z*(-0.03655620 + z*(0.01504268 + z*(-0.00780353 + z*(0.00325614 + z*(-0.00068245)))))))
    return K0, K1

# Тест 3a: сверка с интегральным представлением
for xv in (0.01, 0.5, 1.9, 2.1, 5.0, 20.0):
    tt = np.linspace(0, 12, 200001)
    k0i = trapz(np.exp(-xv*np.cosh(tt)), tt); k1i = trapz(np.cosh(tt)*np.exp(-xv*np.cosh(tt)), tt)
    a_, b_ = K0K1(np.array([xv]))
    check('K0, K1 при x=%.2f' % xv, max(abs(a_[0]/k0i-1), abs(b_[0]/k1i-1)), 2e-7)

# Тест 3b: g0 = K0(kr)/2pi решает (-lap + k^2) g0 = 0 при r != 0 (конечные разности)
kap, r, h = 1.3, 0.7, 1e-3
g = lambda rr: K0K1(np.array([kap*rr]))[0][0]/(2*pi)
lap = (g(r+h) - 2*g(r) + g(r-h))/h**2 + (g(r+h) - g(r-h))/(2*h)/r
check('(-lap+k^2) g0 = 0 при r=0.7', (-lap + kap**2*g(r))/g(r), 1e-5)
# Тест 3c: нормировка — поток -2 pi r g0'(r) = 1 при r -> 0
for rr in (1e-2, 1e-3):
    flux = -2*pi*rr*(g(rr*(1+1e-4)) - g(rr*(1-1e-4)))/(2e-4*rr)
    check('поток -2 pi r dg0/dr = 1 при r=%g' % rr, flux - 1, 2e-3)
```

## 4. Функция Грина при наличии проводников: простой слой

Ищем $g(\mathbf r,\mathbf r')$ — поле в точке $\mathbf r$ от единичного источника в $\mathbf r'$ с $g=0$ на поверхности $\Gamma$ всех проводников. Представим рассеянное поле как потенциал простого слоя — поле зарядов с плотностью $\sigma(\mathbf s)$, сидящих на $\Gamma$:
$$
g(\mathbf r,\mathbf r')=g_0(\mathbf r-\mathbf r')-\int_\Gamma g_0(\mathbf r-\mathbf s)\,\sigma(\mathbf s;\mathbf r')\,dl_s .
$$
Вне $\Gamma$ второе слагаемое удовлетворяет однородному уравнению; на бесконечности убывает. Условие Дирихле при $\mathbf r=\mathbf t\in\Gamma$:
$$
\int_\Gamma g_0(\mathbf t-\mathbf s)\,\sigma(\mathbf s;\mathbf r')\,dl_s = g_0(\mathbf t-\mathbf r')\qquad\Longleftrightarrow\qquad \hat S\sigma=g_0(\cdot-\mathbf r').
$$
$\hat S$ — оператор простого слоя (интегральный оператор с ядром $K_0/2\pi$). Он симметричен и при $\kappa>0$ положительно определён. Отсюда
$$
\boxed{\,g_s(\mathbf r,\mathbf r')=-\big\langle g_0(\mathbf r-\cdot)\,\big|\,\hat S^{-1}\,\big|\,g_0(\cdot-\mathbf r')\big\rangle\,}
$$
Это точная формула для любого числа тел любой формы. Тонкие пластины — частный случай: на них $\Gamma$ — отрезок, и $\sigma$ суммирует заряд обеих сторон. Толстые крылья — замкнутый контур-прямоугольник.

**Физический смысл $\sigma$:** это (с точностью до знака) скачок нормальной производной поля на поверхности, $\sigma=[\partial_n g]$. Внутри замкнутого проводника решение того же уравнения Дирихле — ноль. Поэтому для толстого тела $\sigma$ совпадает с $-\partial_n g$ снаружи.

## 5. Дискретизация (метод панелей)

Контур $\Gamma$ режется на $N$ панелей с серединами $\mathbf m_b$ и длинами $h_b$. Плотность $\sigma$ на каждой панели считается постоянной (кусочно-постоянная коллокация):
$$
\sum_b S_{jb}\,\sigma_b=g_0(\mathbf m_j-\mathbf r'),\qquad S_{jb}=\frac{h_b}{2\pi}K_0(\kappa|\mathbf m_j-\mathbf m_b|)\ \ (j\ne b).
$$
**Диагональ** (самодействие панели) содержит логарифмическую особенность $K_0(x)\approx-\ln(x/2)-\gamma$. Интеграл по панели берётся точно:
$$
S_{jj}=\frac1{2\pi}\int_{-h/2}^{h/2}K_0(\kappa|t|)dt=\frac1\pi\Big[\int_0^{h/2}\big(K_0(\kappa t)+\ln\tfrac{\kappa t}{2}\big)dt-\Big(\tfrac h2\ln\tfrac{\kappa h}{4}-\tfrac h2\Big)\Big].
$$
Гладкую часть $K_0+\ln(\kappa t/2)$ интегрируем 16-точечным Гауссом, $\int\ln$ — аналитически.

**Сгущение узлов.** Около кромок $\sigma$ особенна (для тонкой кромки $\sigma\sim\rho^{-1/2}$, для прямого угла толстого крыла $\sim\rho^{-1/3}$). Поэтому узлы на каждой стороне распределены по косинусу (Чебышёв): $u_n=(1-\cos(\pi n/N))/2$.

В точке наблюдения вне $\Gamma$:
$$
g_s(\mathbf r,\mathbf r')\approx-\sum_b g_0(\mathbf r-\mathbf m_b)\,h_b\,\sigma_b(\mathbf r'),\qquad \sigma(\mathbf r')=S^{-1}g_0(\cdot-\mathbf r').
$$
Нужны производные в совпадающих точках $D_{ij}=\partial_i\partial'_j g_s|_{\mathbf r'=\mathbf r}$. Обе производные переходят на $g_0$:
$$
D_{ij}(\mathbf r)=-\sum_{b,c}\partial_i g_0(\mathbf r-\mathbf m_b)\,h_b\,(S^{-1})_{bc}\,\partial_j g_0(\mathbf m_c-\mathbf r),\qquad
g_s(\mathbf r,\mathbf r)=-\sum_{b,c}g_0\,h_b\,(S^{-1})_{bc}\,g_0 .
$$

```sage
def k0_selfpanel(kap, h):
    """S_jj: (1/2pi) * интеграл K0(kap|t|) по панели длины h (особенность ln вычтена аналитически)."""
    xg, wg = np.polynomial.legendre.leggauss(16)
    T = h/2; t = (xg+1)/2*T; w = wg/2*T
    K0, _ = K0K1(kap*t)
    reg = K0 + np.log(kap*t/2)
    return (np.sum(w*reg) - (T*np.log(kap*T/2) - T)) / pi

def panels(P0, P1, N):
    """N панелей на отрезке P0-P1, сгущение по косинусу к концам. Возвращает (середины, длины)."""
    u = (1 - np.cos(np.linspace(0, pi, N+1))) / 2
    P0 = np.asarray(P0, float); P1 = np.asarray(P1, float)
    nodes = P0[None, :] + u[:, None]*(P1 - P0)[None, :]
    return (nodes[1:] + nodes[:-1])/2, np.linalg.norm(nodes[1:] - nodes[:-1], axis=1)

def rect_wing(P0, P1, t, N, Nt):
    """Крыло толщины t: прямоугольник вдоль P0->P1; N панелей на длинные стороны, Nt на торцы."""
    P0 = np.asarray(P0, float); P1 = np.asarray(P1, float)
    u = (P1-P0)/np.linalg.norm(P1-P0); n = np.array([-u[1], u[0]])
    A, B, C, D = P0-n*t/2, P1-n*t/2, P1+n*t/2, P0+n*t/2
    segs = [panels(A, B, N), panels(B, C, Nt), panels(C, D, N), panels(D, A, Nt)]
    return np.vstack([s[0] for s in segs]), np.concatenate([s[1] for s in segs])

def Smat(mid, h, kap):
    """Матрица простого слоя S_jb."""
    R = np.linalg.norm(mid[:, None, :] - mid[None, :, :], axis=2); np.fill_diagonal(R, 1.0)
    K0, _ = K0K1(kap*R); S = K0/(2*pi)*h[None, :]
    for j in range(len(h)): S[j, j] = k0_selfpanel(kap, h[j])
    return S

def G0grad(r, s, k):
    """g0(r_m - s_b) и его градиент по r: массивы (M,B) и (M,B,2)."""
    d = r[:, None, :] - s[None, :, :]; R = np.linalg.norm(d, axis=2)
    K0, K1 = K0K1(k*R)
    return K0/(2*pi), (-k*K1/(2*pi*R))[..., None]*d

# Тест 5a: самопанель против прямого численного интеграла
kap, hp = 2.0, 0.05
tt = np.linspace(1e-9, hp/2, 400001)
direct = 2*trapz(K0K1(kap*tt)[0], tt)/(2*pi)
check('S_jj против прямой квадратуры', k0_selfpanel(kap, hp)/direct - 1, 1e-5)
```

**Тест 5b: условие Дирихле действительно выполняется.** Источник — в точке $\mathbf r'$ рядом с пластиной. Полное поле $g=g_0+g_s$ в точках *между* серединами панелей (не в узлах коллокации) должно быть мало по сравнению с $g_0$.

```sage
def g_total(bodies, r_obs, r_src, k):
    mid = np.vstack([b[0] for b in bodies]); h = np.concatenate([b[1] for b in bodies])
    S = Smat(mid, h, k)
    Gs_src, _ = G0grad(mid, r_src[None, :], k)          # g0(m_b - r')
    sig = np.linalg.solve(S, Gs_src[:, 0])
    G_obs, _ = G0grad(r_obs, mid, k)                    # g0(r - m_b)
    g0 = G0grad(r_obs, r_src[None, :], k)[0][:, 0]
    return g0, g0 - G_obs @ (h*sig)

plate = panels([-1, 0], [1, 0], 200)
xs = np.linspace(-0.6, 0.6, 7); obs = np.stack([xs, np.zeros_like(xs)], 1) + np.array([0.0013, 0])
g0v, gv = g_total([plate], obs, np.array([0.1, 0.3]), 1.0)
print('g0 на пластине :', np.round(g0v, 4))
print('g  на пластине :', np.round(gv, 5))
check('max|g|/max|g0| на поверхности (Дирихле)', np.max(abs(gv))/np.max(abs(g0v)), 2e-2)
```

## 6. Тензор в точке; закон сохранения $\partial_j\Theta_{ij}=0$

Из §2 и §5 тензор в точке $\mathbf r$ вне проводников:
$$
\langle\Theta_{ij}(\mathbf r)\rangle=\int_0^\infty\frac{\kappa\,d\kappa}{2\pi}\Big[D_{ij}-\tfrac12\delta_{ij}\big(D_{xx}+D_{yy}+\kappa^2 g_s(\mathbf r,\mathbf r)\big)\Big].
$$
**Почему дивергенция равна нулю.** Для каждой $\kappa$ функция $g_s(\mathbf r,\mathbf r')$ решает $(-\nabla^2+\kappa^2)g_s=0$ по каждому аргументу вне $\Gamma$. Возьмём $\partial_j$ от выражения в скобках, используя симметрию $g_s(\mathbf r,\mathbf r')=g_s(\mathbf r',\mathbf r)$:
$$
\partial_j\Big[\partial_i\partial'_jg-\tfrac12\delta_{ij}(\partial_k\partial'_kg+\kappa^2g)\Big]_{\mathbf r'=\mathbf r}
=\big[\partial_i\nabla'^2 g-\kappa^2\partial_i g\big]_{\mathbf r'=\mathbf r}=0 .
$$
(При дифференцировании функции совпадающих точек $f(\mathbf r)=F(\mathbf r,\mathbf r)$ получаем $\partial_jf=(\partial_j+\partial'_j)F$; перекрёстные члены сокращаются.) Значит, для каждого $\kappa$ отдельно $\partial_j\Theta_{ij}=0$ — это и проверяется численно ниже.

**Интеграл по $\kappa$.** $g_s$ убывает примерно как $e^{-2\kappa d}$, где $d$ — расстояние от точки до ближайшего проводника. Поэтому замена $\kappa=s\tan(\pi u/2)$, $u\in(0,1)$, с масштабом $s\sim1/d$ и средними точками по $u$ сходится быстро; 50–60 узлов хватает.

```sage
def kgrid(n=60, s=1.0):
    """Узлы и веса для интеграла по kappa от 0 до бесконечности: kappa = s*tan(pi u/2)."""
    u = (np.arange(n) + 0.5)/n
    return s*np.tan(pi/2*u), s*pi/2/np.cos(pi/2*u)**2/n

def tensor(bodies, pts, nk=60, smin=None):
    """<Theta_ij>(r) в точках pts (M,2), интегрированный по kappa. Возвращает (M,2,2)."""
    mid = np.vstack([b[0] for b in bodies]); h = np.concatenate([b[1] for b in bodies])
    dmin = np.min(np.linalg.norm(pts[:, None, :] - mid[None, :, :], axis=2))
    kk, ww = kgrid(nk, smin if smin else 1.0/dmin)
    Th = np.zeros((len(pts), 2, 2))
    for k, w in zip(kk, ww):
        S = Smat(mid, h, k); G, gr = G0grad(pts, mid, k)     # g0(r-m_b), grad_r g0(r-m_b)
        X0 = np.linalg.solve(S, G.T)                          # S^{-1} g0(m-r)
        Xg = np.stack([np.linalg.solve(S, gr[:, :, j].T) for j in range(2)], -1)
        # grad_{r'} g0(m-r') = +grad_r g0(r-m)|_{r=r'} * (-1) * (-1): g0 зависит от |r-m|, d/dr' g0(m-r') = -d/dm = grad_r g0(r-m)
        gs = -np.einsum('mb,bm->m', G*h[None, :], X0)
        D = -np.einsum('mbi,bmj->mij', gr*h[None, :, None], Xg)
        Th += w*k/(2*pi)*(D - 0.5*np.eye(2)[None]*(k*k*gs + D[:, 0, 0] + D[:, 1, 1])[:, None, None])
    return Th

def traction(bodies, pts, nrm, nk=60, smin=None):
    """Плотность силы на единицу длины контура: sigma_ij n_j = -Theta_ij n_j."""
    return -np.einsum('mij,mj->mi', tensor(bodies, pts, nk, smin), nrm)
```

**Замечание о знаке производной по $\mathbf r'$.** $g_0$ зависит только от $|\mathbf r-\mathbf m|$. Поэтому $\partial'_j g_0(\mathbf m-\mathbf r')$ — производная по второму аргументу — равна $\partial_j g_0(\mathbf r-\mathbf m)$ при $\mathbf r=\mathbf r'$. Две смены знака (от $\mathbf m-\mathbf r'$ и от чётности) взаимно уничтожаются, поэтому в коде для обоих множителей стоит один и тот же массив `gr`.

**Тест 6: $\partial_j\Theta_{ij}=0$ численно.** Две пластины под углом, точка в зазоре. Центральные разности с шагом $\delta$; сравниваем с характерным масштабом $|\Theta|/\ell$.

```sage
A = panels([-0.5, 0], [0.5, 0], 80); B = panels([-0.4, 0.5], [0.6, 0.35], 80)
p0 = np.array([0.05, 0.22]); dlt = 2e-3
pts = np.array([p0 + [dlt, 0], p0 - [dlt, 0], p0 + [0, dlt], p0 - [0, dlt], p0])
Th = tensor([A, B], pts, nk=60, smin=1/0.2)
div = (Th[0][:, 0] - Th[1][:, 0])/(2*dlt) + (Th[2][:, 1] - Th[3][:, 1])/(2*dlt)
scale = np.max(abs(Th[4]))/0.2
print('Theta в точке:\n', Th[4]); print('div Theta =', div, '  масштаб |Theta|/l =', scale)
check('|div Theta| / (|Theta|/l)', np.max(abs(div))/scale, 1e-3)
```

## 7. Сила как поток тензора; деформация контура

Возьмём область $V$, ограниченную снаружи контуром $C$, внутри — поверхностью тела 1. Других тел внутри нет. В $V$ $\partial_j\sigma_{ij}=0$, поэтому по теореме Гаусса
$$
\oint_C\sigma_{ij}n_j\,dl=\oint_{\Gamma_1^+}\sigma_{ij}n_j\,dl ,
$$
где $\Gamma_1^+$ — сторона поверхности тела, обращённая наружу. Внутри идеального проводника поля нет, и правая часть — это по определению полная сила на тело 1 (давление поля на его поверхность). Отсюда:

* **Любой** контур, охватывающий тело 1 и не задевающий другие тела, даёт **одну и ту же** силу — от большой окружности до контура, прижатого к металлу на расстояние $\varepsilon\to0$.
* **Вариант 1** (окружность вокруг всего уголка) даёт полную силу на уголок как целое. Чтобы получить силу на одно крыло, окружность не годится: у вершины зазор между крыльями всего 0,075, поэтому контур должен обходить крыло вплотную.
* **Вариант 2** (контур, обжатый вокруг каждого крыла) даёт силу на каждое крыло и её раскладку по граням и торцам. Сумма по двум крыльям — та же полная сила, что в варианте 1.

**Чего вариант 2 не может дать.** Сила на *часть* поверхности (грань, торец) — не наблюдаемая величина. При $\varepsilon\to0$ контур упирается в углы прямоугольника, где поле особенно (угол $3\pi/2$: $\varphi\sim\rho^{2/3}$, $\Theta\sim\rho^{-2/3}$). Кроме того, на расстояниях $\varepsilon$, сравнимых с размером панели, контур «видит» дискретность простого слоя. Поэтому вклады отдельных частей зависят от $\varepsilon$ и от сетки панелей, и при малых $\varepsilon$ растут. Конечна и устойчива только их сумма по замкнутому контуру. В 10.2 это видно напрямую, а в 10.4 сделана отдельная проверка на сгущение сетки.

**Знак.** $F_i=-\oint\Theta_{ij}n_j\,dl$, $n$ — наружу от тела. Знак проверяется независимо: (а) параллельные пластины притягиваются; (б) совпадение с энергетическим методом в §9.

## 8. Энергетический метод (Ллойд / TGTG) — независимая проверка

Энергия Казимира на единицу длины (ТМ, Дирихле):
$$
E=\frac{1}{4\pi}\int_0^\infty\kappa\,d\kappa\;\ln\frac{\det \hat S_{12}}{\det\hat S_1\,\det\hat S_2}.
$$
$\hat S_{12}$ — оператор простого слоя на объединении границ, $\hat S_{1,2}$ — на каждой по отдельности.

**Откуда формула.** Условие $\varphi|_\Gamma=0$ вводится в функциональный интеграл через множитель Лагранжа $\lambda(\mathbf s)$ на $\Gamma$: $\delta(\varphi|_\Gamma)=\int D\lambda\,e^{i\int\lambda\varphi}$. Интеграл по $\varphi$ гауссов и даёт $\exp(-\tfrac12\langle\lambda|\hat S|\lambda\rangle)$. Затем интеграл по $\lambda$ даёт $(\det\hat S)^{-1/2}$. Свободная энергия $=\tfrac12\ln\det\hat S$ на каждую пару $(\xi,k_z)$; мера $\int d\xi\,dk_z/(2\pi)^2\to\int\kappa d\kappa/2\pi$. Отношение определителей убирает расходящиеся самоэнергии тел: остаётся конечная энергия взаимодействия, $\ln\det(1-T_1g_0T_2g_0)$ в обозначениях TGTG.

**Сила** на тело 1 при жёстком сдвиге: $F=-\partial E/\partial X_1$. При дискретизации удобно взять производную по положению каждой панели:
$$
\frac{\partial}{\partial\mathbf m_j}\ln\det S=\mathrm{Tr}\Big(S^{-1}\frac{\partial S}{\partial\mathbf m_j}\Big),
$$
где $\partial S/\partial\mathbf m_j$ отлична от нуля лишь в $j$-й строке и $j$-м столбце (диагональ от сдвига не зависит). Сумма по панелям тела 1 — сила на тело 1.

Это **совершенно другой** путь: нет тензора, нет контуров, нет совпадающих точек. Совпадение двух методов проверяет и нормировку, и знак, и дискретизацию.

```sage
def grad_lndet(mid, h, k):
    """d ln det S / d m_j для всех панелей j: массив (N,2), и ln det S."""
    S = Smat(mid, h, k); Si = np.linalg.inv(S)
    d = mid[:, None, :] - mid[None, :, :]; R = np.linalg.norm(d, axis=2); np.fill_diagonal(R, 1.0)
    _, K1 = K0K1(k*R); np.fill_diagonal(K1, 0.0)
    g = (k*K1/(2*pi*R))[..., None]*d               # = -grad_{m_i} K0/2pi (i-я строка)
    G = np.einsum('ji,ijc->jc', Si, g)*h[:, None] + np.einsum('ij,ijc,i->jc', Si, g, h)
    return G, np.linalg.slogdet(S)[1]

def casimir(W1, W2, nk=60, ks=3.0):
    """Энергия взаимодействия и силы на панели (интеграл по kappa). Возвращает E, F(панели), N1."""
    (m1, h1), (m2, h2) = W1, W2
    mid = np.vstack([m1, m2]); h = np.concatenate([h1, h2]); N1 = len(h1)
    kk, ww = kgrid(nk, ks); F = np.zeros((len(h), 2)); E = 0.0
    for k, w in zip(kk, ww):
        G, L = grad_lndet(mid, h, k); G1, L1 = grad_lndet(m1, h1, k); G2, L2 = grad_lndet(m2, h2, k)
        G[:N1] -= G1; G[N1:] -= G2
        F += -(1/(4*pi))*w*k*G
        E += w*k*(L - L1 - L2)
    return E/(4*pi), F, N1

# Тест 8a: F = -dE/dX численным сдвигом тела 2
A = panels([-0.5, 0], [0.5, 0], 60); B = panels([1.5, -0.3], [1.9, 0.6], 40)
E0, Fv, NA = casimir(A, B, nk=50)
dx = 1e-3
Ep, _, _ = casimir(A, (B[0] + [dx, 0], B[1]), nk=50); Em, _, _ = casimir(A, (B[0] - [dx, 0], B[1]), nk=50)
FB_num = -(Ep - Em)/(2*dx); FB = Fv[NA:].sum(0)
print('E = %.6e;  F_B(след) = %+.6e, -dE/dX = %+.6e' % (E0, FB[0], FB_num))
check('сила через Tr(S^-1 dS) = -dE/dX', FB[0]/FB_num - 1, 1e-4)
check('третий закон: F_A + F_B = 0', np.linalg.norm(Fv.sum(0))/np.linalg.norm(FB), 1e-8)
```

## 9. Тест метода тензора на двух разнесённых телах

Окружность вокруг тела A (не задевающая B), радиусы 0,8 и 1,0. Поток тензора должен совпасть с силой на A из энергетического метода и не зависеть от радиуса.

```sage
def circle_force(bodies, center, Rc, M=300, nk=60):
    th = 2*pi*np.arange(M)/M; n = np.stack([np.cos(th), np.sin(th)], 1)
    pts = center + Rc*n
    return (traction(bodies, pts, n, nk) * (2*pi*Rc/M)).sum(0)

FA = Fv[:NA].sum(0)
print('энергетический метод, сила на A: Fx = %+.6e  Fy = %+.6e' % tuple(FA))
for Rc in (0.8, 1.0):
    Ft = circle_force([A, B], np.array([0.0, 0.0]), Rc, nk=50)
    print('поток тензора, R=%.1f:           Fx = %+.6e  Fy = %+.6e' % ((Rc,) + tuple(Ft)))
    check('поток тензора = сила на A (R=%.1f)' % Rc, np.linalg.norm(Ft - FA)/np.linalg.norm(FA), 2e-3)
```

## 10. Уголок

Два прямоугольных крыла толщины $t=0{,}08$ вдоль лучей под углами $\pm\alpha$ к оси $y$, $\alpha=15^\circ$. Крылья занимают $r\in[r_0,r_1]=[0{,}3;\,3]$ от вершины. Зазор между внутренними гранями у вершины: $2r_0\sin\alpha-t\approx0{,}075$. По 140 панелей на длинных сторонах, 16 на торцах.

### 10.0. Эталон: энергетический метод
Сила на каждое крыло и сумма. Если однотельная тяга есть, она проявится как $F_y^{(1)}+F_y^{(2)}\neq0$.

```sage
a = np.radians(15); r0, r1, t = 0.3, 3.0, 0.08
u1 = np.array([np.sin(a), np.cos(a)]); u2 = np.array([-np.sin(a), np.cos(a)])
W1 = rect_wing(r0*u1, r1*u1, t, 140, 16); W2 = rect_wing(r0*u2, r1*u2, t, 140, 16)
print('зазор у вершины:', 2*r0*np.sin(a) - t, '   панелей:', len(W1[1]) + len(W2[1]))
t0 = time.time()
E, Fv, N1 = casimir(W1, W2, nk=60, ks=3.0)
F1E, F2E = Fv[:N1].sum(0), Fv[N1:].sum(0)
print('E = %.6f' % E)
print('крыло 1: F = (%+.6f, %+.3e)' % tuple(F1E)); print('крыло 2: F = (%+.6f, %+.3e)' % tuple(F2E))
print('сумма  : F = (%+.3e, %+.3e)   [%.0f c]' % (tuple(F1E + F2E) + (time.time() - t0,)))
```

### 10.2. Вариант 2: контур, обжатый вокруг крыла

Контур идёт параллельно граням на расстоянии $\varepsilon$ от металла, а у четырёх углов прямоугольника — по четвертям окружностей радиуса $\varepsilon$. Нормаль направлена наружу от крыла. На каждой прямой стороне — составная квадратура Гаусса–Лежандра с геометрическим сгущением к углам (первая ячейка $\sim\varepsilon/2$), на дугах — 16 точек.

Части контура:
* `side-`, `side+` — две длинные грани (у крыла 1 `side+` смотрит на крыло 2);
* `tip` — свободный торец со скруглениями;
* `vertex` — торец у вершины уголка.

Для каждой части печатается $\oint\sigma_{ij}n_j\,dl$ по этой части, для крыла — сумма, ниже — сумма по двум крыльям.

```sage
def graded(L, eps, nq=8, ratio=2.0):
    """Составная квадратура Гаусса на [0,L] с геометрическим сгущением к обоим концам."""
    edges = [0.0]; x = eps/2
    while x < L/2: edges.append(x); x *= ratio
    half = np.array(edges + [L/2]); allE = np.concatenate([half, L - half[::-1][1:]])
    xg, wg = np.polynomial.legendre.leggauss(nq); s = []; w = []
    for a_, b_ in zip(allE[:-1], allE[1:]):
        s.extend(a_ + (xg+1)/2*(b_-a_)); w.extend(wg/2*(b_-a_))
    return np.array(s), np.array(w)

def hug(P0, P1, t, eps, nq=8, narc=16):
    """Контур на расстоянии eps вокруг крыла P0->P1 толщины t: точки, нормали, веса dl, метки частей."""
    P0 = np.asarray(P0, float); P1 = np.asarray(P1, float); L = np.linalg.norm(P1-P0)
    u = (P1-P0)/L; n = np.array([-u[1], u[0]]); hw = t/2
    P = []; N_ = []; DL = []; LB = []
    def line(A, B, nn, lab):
        Ls = np.linalg.norm(B-A); s, w = graded(Ls, eps, nq)
        P.extend(A + np.outer(s/Ls, B-A)); N_.extend([nn]*len(s)); DL.extend(w); LB.extend([lab]*len(s))
    def arc(C, th0, th1, lab):
        xg, wg = np.polynomial.legendre.leggauss(narc); th = th0 + (xg+1)/2*(th1-th0)
        d = np.cos(th)[:, None]*u + np.sin(th)[:, None]*n
        P.extend(C + eps*d); N_.extend(d); DL.extend(eps*wg/2*(th1-th0)); LB.extend([lab]*narc)
    line(P0-n*(hw+eps), P1-n*(hw+eps), -n, 'side-')
    arc(P1-n*hw, -pi/2, 0, 'tip'); line(P1+u*eps-n*hw, P1+u*eps+n*hw, u, 'tip'); arc(P1+n*hw, 0, pi/2, 'tip')
    line(P1+n*(hw+eps), P0+n*(hw+eps), n, 'side+')
    arc(P0+n*hw, pi/2, pi, 'vertex'); line(P0-u*eps+n*hw, P0-u*eps-n*hw, -u, 'vertex'); arc(P0-n*hw, pi, 1.5*pi, 'vertex')
    return np.array(P), np.array(N_), np.array(DL), np.array(LB)

parts = ('side-', 'side+', 'tip', 'vertex')
surf = {}
for eps in (0.04, 0.02, 0.01):
    t0 = time.time(); tot = []
    for (P0, P1) in ((r0*u1, r1*u1), (r0*u2, r1*u2)):
        p, nn, dl, lb = hug(P0, P1, t, eps)
        Tr = traction([W1, W2], p, nn, nk=60, smin=1/eps)
        Fp = {L: (Tr[lb == L]*dl[lb == L, None]).sum(0) for L in parts}
        Fp['total'] = (Tr*dl[:, None]).sum(0); tot.append(Fp)
    surf[eps] = tot
    print('eps = %.3f   [%.0f c]' % (eps, time.time()-t0))
    for L in parts + ('total',):
        print('   %-7s крыло1 (%+11.4f, %+11.4f)   крыло2 (%+11.4f, %+11.4f)' % ((L,) + tuple(tot[0][L]) + tuple(tot[1][L])))
    S = tot[0]['total'] + tot[1]['total']
    print('   сумма по двум крыльям: (%+.2e, %+.2e);   |F|/|F_x^(1)| = %.1e' % (S[0], S[1], np.linalg.norm(S)/abs(F1E[0])))
```

Тот же контур $\varepsilon=0{,}04$ вокруг крыла 1, но тела дискретизованы гуще (200 панелей на длинных сторонах, 24 на торцах вместо 140/16). Сумма должна остаться прежней, части — нет.

## 11. ТЕ-поляризация: условие Неймана

Для ТЕ ($E_z=0$) всё поле выражается через $\psi=H_z$, а условие на идеальном проводнике — $\partial_n\psi|_\Gamma=0$ (касательное $E\propto\hat z\times\nabla\psi$ обращается в ноль). Тензор $\Theta_{ij}$ и формула §2 те же самые, меняется только функция Грина.

**Представление.** Снова простой слой: $g_s(\mathbf r,\mathbf r')=\int_\Gamma g_0(\mathbf r-\mathbf s)\,\mu(\mathbf s;\mathbf r')\,dl_s$, но теперь на $\Gamma$ требуем
$\partial_n(g_0+g_s)=0$. Нормальная производная простого слоя при подходе снаружи имеет скачок (формула Племеля–Привалова):
$$
\partial_n g_s\big|_{\Gamma^+}=\Big(-\tfrac12\hat I+\hat K'\Big)\mu,\qquad (\hat K'\mu)(\mathbf t)=\mathrm{v.p.}\!\int_\Gamma \partial_{n_t}g_0(\mathbf t-\mathbf s)\,\mu(\mathbf s)\,dl_s .
$$
Получаем интегральное уравнение второго рода
$$
\Big(-\tfrac12\hat I+\hat K'\Big)\mu=-\partial_{n}g_0(\cdot-\mathbf r'),\qquad
g_s(\mathbf r,\mathbf r')=\big\langle g_0(\mathbf r-\cdot)\,\big|\,\hat A^{-1}\,\big|\,b_{\mathbf r'}\big\rangle,\ \ \hat A=-\tfrac12+\hat K',\ b_{\mathbf r'}=-\partial_n g_0(\cdot-\mathbf r').
$$
(Знак $-\tfrac12$ — для нормали наружу от тела: $\partial_n$ потенциала простого слоя испытывает скачок $-\mu$ при переходе изнутри наружу.) Ядро $\partial_n g_0$ на гладком куске контура гладкое; на прямой панели оно равно нулю для самой панели, поэтому $K'_{jj}=0$.

**Производные в совпадающих точках:** $D_{ij}=\sum_{b,c}\partial_ig_0(\mathbf r-\mathbf m_b)h_b(A^{-1})_{bc}\,\partial'_j b_c(\mathbf r')$, где $\partial'_j b_c=-\partial_j(\mathbf n_c\cdot\nabla g_0)$ — вторые производные (гессиан) $g_0$:
$$
\partial_i\partial_jg_0=\Big(g_0''-\frac{g_0'}{R}\Big)e_ie_j+\frac{g_0'}{R}\delta_{ij},\quad g_0'=-\frac{\kappa K_1}{2\pi},\ \ g_0''=\frac{\kappa^2}{2\pi}\Big(K_0+\frac{K_1}{\kappa R}\Big).
$$

**Энергия (Ллойд для Неймана):** $E=\frac1{4\pi}\int\kappa d\kappa\,\ln\frac{\det A_{12}}{\det A_1\det A_2}$. Отношение определителей второго рода корректно: $-\tfrac12\hat I+\hat K'$ отличается от одночастичных на компактный оператор связи. Силу считаем численной производной $E$ по сдвигу крыла.

**Точность.** Неймановский решатель здесь заметно грубее дирихлевского. Тест на круге: $\hat A\cdot1$ даёт $-0{,}87956$ против точного $-\kappa a I_0K_1=-0{,}88080$ (0,14 %). На параллельных пластинах давление получается в пределах ~4 % от идеального (см. тест). Вблизи панелей (ближе их размера) поле уже не точное. Поэтому полную силу на уголок надёжнее брать по окружности и энергией.

```sage
def poly_body(V, Ns):
    """Многоугольник с вершинами V (против часовой), Ns панелей на сторонах: (середины, длины, внешние нормали)."""
    M = []; H = []; NN = []
    for i in range(len(V)):
        X = np.asarray(V[i], float); Y = np.asarray(V[(i+1) % len(V)], float)
        m, h = panels(X, Y, Ns[i]); tau = (Y-X)/np.linalg.norm(Y-X)
        M.append(m); H.append(h); NN.append(np.tile([tau[1], -tau[0]], (Ns[i], 1)))
    return np.vstack(M), np.concatenate(H), np.vstack(NN)

def rect_wing_n(P0, P1, t, N, Nt):
    """То же крыло, что rect_wing, но с нормалями (обход против часовой)."""
    P0 = np.asarray(P0, float); P1 = np.asarray(P1, float)
    u = (P1-P0)/np.linalg.norm(P1-P0); n = np.array([-u[1], u[0]])
    return poly_body([P0-n*t/2, P1-n*t/2, P1+n*t/2, P0+n*t/2], [N, Nt, N, Nt])

def G0hess(r, s, k):
    d = r[:, None, :] - s[None, :, :]; R = np.linalg.norm(d, axis=2)
    K0, K1 = K0K1(k*R); e = d/R[..., None]
    g1 = -k*K1/(2*pi); g2 = k*k/(2*pi)*(K0 + K1/(k*R))
    return (g2 - g1/R)[..., None, None]*e[..., :, None]*e[..., None, :] + (g1/R)[..., None, None]*np.eye(2)

def Nmat(mid, h, nrm, k):
    """A = -I/2 + K',  K'_jb = h_b n_j . grad g0(m_j - m_b),  K'_jj = 0 (прямая панель)."""
    d = mid[:, None, :] - mid[None, :, :]; R = np.linalg.norm(d, axis=2); np.fill_diagonal(R, 1.0)
    _, K1 = K0K1(k*R); gr = (-k*K1/(2*pi*R))[..., None]*d
    Kp = np.einsum('jbc,jc->jb', gr, nrm)*h[None, :]; np.fill_diagonal(Kp, 0.0)
    return -0.5*np.eye(len(h)) + Kp

def tensorN(bodies, pts, nk=60, smin=None):
    """<Theta_ij> для условия Неймана (ТЕ)."""
    mid = np.vstack([b[0] for b in bodies]); h = np.concatenate([b[1] for b in bodies]); nrm = np.vstack([b[2] for b in bodies])
    dmin = np.min(np.linalg.norm(pts[:, None, :] - mid[None, :, :], axis=2))
    kk, ww = kgrid(nk, smin if smin else 1.0/dmin); Th = np.zeros((len(pts), 2, 2))
    for k, w in zip(kk, ww):
        A = Nmat(mid, h, nrm, k); G, gr = G0grad(pts, mid, k); Hs = G0hess(pts, mid, k)
        Bv = np.einsum('mbi,bi->bm', gr, nrm)          # b_c(r') = -dn_c g0(m_c - r') = n_c . grad_r g0(r - m_c)
        Bd = np.einsum('mbij,bi->bmj', Hs, nrm)        # d/dr'_j  b_c(r')
        X0 = np.linalg.solve(A, Bv); Xd = np.stack([np.linalg.solve(A, Bd[:, :, j]) for j in range(2)], -1)
        gs = np.einsum('mb,bm->m', G*h[None, :], X0)
        D = np.einsum('mbi,bmj->mij', gr*h[None, :, None], Xd)
        Th += w*k/(2*pi)*(D - 0.5*np.eye(2)[None]*(k*k*gs + D[:, 0, 0] + D[:, 1, 1])[:, None, None])
    return Th

def tractionN(bodies, pts, nrm, nk=60, smin=None):
    return -np.einsum('mij,mj->mi', tensorN(bodies, pts, nk, smin), nrm)

def energyN(bodies, nk=60, ks=3.0):
    mid = np.vstack([b[0] for b in bodies]); h = np.concatenate([b[1] for b in bodies]); nrm = np.vstack([b[2] for b in bodies])
    kk, ww = kgrid(nk, ks); E = 0.0
    for k, w in zip(kk, ww):
        E += w*k*(np.linalg.slogdet(Nmat(mid, h, nrm, k))[1] - sum(np.linalg.slogdet(Nmat(b[0], b[1], b[2], k))[1] for b in bodies))
    return E/(4*pi)

# Тест 11a: оператор A на круге радиуса a, плотность 1: точно -kappa a I0(kappa a) K1(kappa a)
from math import factorial
k, a0 = 1.0, 0.5
th = np.linspace(0, 2*pi, 401); Vc = np.stack([np.cos(th[:-1]), np.sin(th[:-1])], 1)*a0
mc, hc, nc = poly_body(Vc, [1]*400)
x = k*a0; I0 = sum((x/2)**(2*m)/factorial(m)**2 for m in range(30)); K1v = K0K1(np.array([x]))[1][0]
check('Нейман: A*1 на круге = -ka I0 K1', (Nmat(mc, hc, nc, k) @ np.ones(400))[0]/(-x*I0*K1v) - 1, 3e-3)

# Тест 11b: давление между толстыми пластинами 6 x 0.1, зазор 0.3; идеал для каждой поляризации pi^2/(480 d^4)
d = 0.3; Lp = 6.0; tt = 0.1
P1 = poly_body([[-Lp/2, -tt], [Lp/2, -tt], [Lp/2, 0], [-Lp/2, 0]], [120, 10, 120, 10])
P2 = poly_body([[-Lp/2, d], [Lp/2, d], [Lp/2, d+tt], [-Lp/2, d+tt]], [120, 10, 120, 10])
pt = np.array([[0.0, d/2]]); ideal = pi**2/480/d**4
pD = -tensor([P1[:2], P2[:2]], pt, nk=60, smin=1/d)[0, 1, 1]; pN = -tensorN([P1, P2], pt, nk=60, smin=1/d)[0, 1, 1]
print('давление в центре зазора: Дирихле %.4f, Нейман %.4f, идеал %.4f' % (pD, pN, ideal))
check('Дирихле: p / идеал - 1 (конечная длина пластин)', pD/ideal - 1, 5e-2)
check('Нейман:  p / идеал - 1 (конечная длина пластин)', pN/ideal - 1, 5e-2)
```

## 12. Дополнительные проверки: «прозрачный» металл и почти сомкнутая вершина

**Полупрозрачные крылья.** Идеальный проводник заменяем тонким слоем с потенциалом $V=\lambda\,\delta_\Gamma$: $(-\nabla^2+\kappa^2+\lambda\delta_\Gamma)g=\delta$. Это стандартная модель Бартона/Милтона «плазменной плёнки»: при $\lambda\to\infty$ получается Дирихле, при малых $\lambda$ поле частично проходит сквозь металл (аналог конечной глубины проникновения, «предела прозрачности»). В методе простого слоя меняется одно: $\hat S\to\hat S+\hat I/\lambda$, так как скачок $\partial_n g$ теперь равен $\lambda g$, а не произвольному $\sigma$ при $g=0$. Тензор вне слоя и все формулы §2–§8 остаются прежними. Матрица $1/\lambda$ от положения панелей не зависит, поэтому формула силы через след тоже не меняется.

**Почти сомкнутая вершина.** Тонкие крылья ($t\to0$) от $r_0=0{,}3$ и от $r_0=0{,}05$ до $r_1=3$, 220 панелей на крыло. При $r_0=0{,}05$ щель у вершины $2r_0\sin15^\circ=0{,}026$, сила сжатия в 200 раз больше. Если однотельная сила рождается у вершины, здесь она проявилась бы сильнее всего.

Для каждого случая — сила на крыло 1 энергией, сумма сил на два крыла и поток тензора по окружности $R=2{,}2$.

```sage
def Smat_l(mid, h, k, lam):
    S = Smat(mid, h, k)
    return S if lam is None else S + np.eye(len(h))/lam

def circle_force_l(bodies, center, Rc, lam, M=400, nk=60):
    mid = np.vstack([b[0] for b in bodies]); h = np.concatenate([b[1] for b in bodies])
    th = 2*pi*np.arange(M)/M; n = np.stack([np.cos(th), np.sin(th)], 1); pts = center + Rc*n
    dmin = np.min(np.linalg.norm(pts[:, None, :] - mid[None, :, :], axis=2)); kk, ww = kgrid(nk, 1/dmin); F = np.zeros(2)
    for k, w in zip(kk, ww):
        S = Smat_l(mid, h, k, lam); Gm, gr = G0grad(pts, mid, k)
        X0 = np.linalg.solve(S, Gm.T); Xg = np.stack([np.linalg.solve(S, gr[:, :, j].T) for j in range(2)], -1)
        gs = -np.einsum('mb,bm->m', Gm*h[None, :], X0); D = -np.einsum('mbi,bmj->mij', gr*h[None, :, None], Xg)
        Th = D - 0.5*np.eye(2)[None]*(k*k*gs + D[:, 0, 0] + D[:, 1, 1])[:, None, None]
        F += w*k/(2*pi)*np.einsum('mij,mj->i', Th, n)*(2*pi*Rc/M)
    return -F

def force_l(W1, W2, lam, nk=60, ks=3.0):
    (m1, h1), (m2, h2) = W1, W2; mid = np.vstack([m1, m2]); h = np.concatenate([h1, h2]); N1 = len(h1)
    def gl(mid, h, k):
        Si = np.linalg.inv(Smat_l(mid, h, k, lam))
        d = mid[:, None, :] - mid[None, :, :]; R = np.linalg.norm(d, axis=2); np.fill_diagonal(R, 1.0)
        _, K1 = K0K1(k*R); np.fill_diagonal(K1, 0.0); g = (k*K1/(2*pi*R))[..., None]*d
        return np.einsum('ji,ijc->jc', Si, g)*h[:, None] + np.einsum('ij,ijc,i->jc', Si, g, h)
    kk, ww = kgrid(nk, ks); F = np.zeros((len(h), 2))
    for k, w in zip(kk, ww):
        G_ = gl(mid, h, k); G_[:N1] -= gl(m1, h1, k); G_[N1:] -= gl(m2, h2, k); F += -(1/(4*pi))*w*k*G_
    return F[:N1].sum(0), F[N1:].sum(0)

for r0_ in (0.3, 0.05):
    A1 = panels(r0_*u1, 3*u1, 220); A2 = panels(r0_*u2, 3*u2, 220)
    for lam in (1.0, 10.0, 100.0, None):
        F1, F2 = force_l(A1, A2, lam); Fc = circle_force_l([A1, A2], cen, 2.2, lam)
        print('r0=%.2f  lambda=%-5s  F1=(%+.4e, %+.0e)  F1+F2=(%+.0e, %+.0e)  окружность |F|/|F1x| = %.0e'
              % (r0_, lam, F1[0], F1[1], *(F1 + F2), np.linalg.norm(Fc)/abs(F1[0])))
```

Прозрачность меняет силу сжатия на два порядка, сближение вершины — ещё на два с половиной. Однотельная сила остаётся нулём на уровне машинной точности энергетического метода.

**Почему так — и где граница вывода.** Для статического равновесного поля (любая $\lambda$, любая форма, ТМ и ТЕ, а также $T>0$ — там $\int d\xi$ заменяется суммой Мацубары, но каждое слагаемое имеет ту же структуру) энергия $E(\{X\})$ зависит только от *взаимного* положения тел. Сдвиг всего уголка как целого её не меняет, а значит $\sum_jF_j=-\partial E/\partial X_{\rm общ}=0$. Это трансляционная инвариантность (Нётер) в применении к свободной энергии. Тензорный ноль по окружности — то же самое, выраженное через Гаусса. Кроме того, в статике плотность импульса поля $\mathbf g=0$, поэтому вакууму «в щели» некуда принять противоположный импульс.

Под этот аргумент **не попадают**: (а) неравновесные состояния (тела при разных температурах — известны однотельные «тепловые» силы на асимметричные тела, если тело теплее окружения: оно излучает анизотропно; это фотонная ракета, $|F|\le P/c$); (б) переменные во времени поля и движущиеся границы (динамический Казимир, там $\partial_t\mathbf g\ne0$); (в) нелинейные/невзаимные среды.

## 12b. Только тензор: где именно приложена сила (тонкие крылья)

Энергия здесь не используется. На каждом крыле берём «стадион» — контур на расстоянии $\rho$ от тонкой пластины: две прямые (внутренняя грань, обращённая в щель, и внешняя) и две полуокружности (торец у вершины и свободный торец).

**Ключевой шаг — тензор взаимодействия.** Вычитаем из тензора системы тензоры каждого крыла в одиночестве:
$$
\Theta^{\rm int}_{ij}=\Theta_{ij}[1+2]-\Theta_{ij}[1]-\Theta_{ij}[2].
$$
Тензор одиночного крыла сингулярен у его кромок (именно отсюда огромные числа по частям в §10.2). Но его полный поток по любому контуру вокруг крыла равен нулю (одиночное тело само себя не толкает), а по контуру вокруг другого крыла — нулю по Гауссу. Поэтому $\Theta^{\rm int}$ даёт ту же полную силу. Вклады отдельных частей контура становятся конечными и физически осмысленными: давление поля *взаимодействия* на каждую грань.

Затем экстраполируем к $\rho\to0$ (линейно по $\rho$).

```sage

def tensor_l(bodies, pts, lam, nk=50, smin=None, chunk=150):
    mid=np.vstack([b[0] for b in bodies]); h=np.concatenate([b[1] for b in bodies])
    dmin=np.min(np.linalg.norm(pts[:,None,:]-mid[None,:,:],axis=2)); kk,ww=kgrid(nk,smin if smin else 1/dmin)
    Th=np.zeros((len(pts),2,2))
    for k,w in zip(kk,ww):
        Si=np.linalg.inv(Smat_l(mid,h,k,lam))
        for i0 in range(0,len(pts),chunk):
            q=pts[i0:i0+chunk]; Gm,gr=G0grad(q,mid,k)
            X0=Si@Gm.T; Xg=np.einsum('bc,mcj->bmj',Si,gr)
            gs=-np.einsum('mb,bm->m',Gm*h[None,:],X0); D=-np.einsum('mbi,bmj->mij',gr*h[None,:,None],Xg)
            Th[i0:i0+chunk]+=w*k/(2*pi)*(D-0.5*np.eye(2)[None]*(k*k*gs+D[:,0,0]+D[:,1,1])[:,None,None])
    return Th
def stadium(P0,P1,rho,nq=6,narc=24):
    P0=np.asarray(P0,float);P1=np.asarray(P1,float);L=np.linalg.norm(P1-P0);u=(P1-P0)/L;n=np.array([-u[1],u[0]])
    P=[];N_=[];DL=[];LB=[]
    s,w=graded(L,rho,nq)
    for sg,lab in ((-1,'outer'),(1,'inner')):
        P.extend(P0+sg*rho*n+np.outer(s,u)); N_.extend([sg*n]*len(s)); DL.extend(w); LB.extend([lab]*len(s))
    xg,wg=np.polynomial.legendre.leggauss(narc)
    for C,th0,lab in ((P1,-pi/2,'tip'),(P0,pi/2,'vertex')):
        th=th0+(xg+1)/2*pi; d=np.cos(th)[:,None]*u+np.sin(th)[:,None]*n
        P.extend(C+rho*d); N_.extend(d); DL.extend(rho*wg/2*pi); LB.extend([lab]*narc)
    return np.array(P),np.array(N_),np.array(DL),np.array(LB)
def wing_parts(bodies,P0,P1,rho,lam):
    p,nn,dl,lb=stadium(P0,P1,rho)
    Tr=-np.einsum('mij,mj->mi',tensor_l(bodies,p,lam,smin=1/rho),nn)
    F={L:(Tr[lb==L]*dl[lb==L,None]).sum(0) for L in ('inner','outer','vertex','tip')}; F['wing']=(Tr*dl[:,None]).sum(0)
    return F,len(p)

def wing_parts_int(A1, A2, P0, P1, rho, lam):
    p, nn, dl, lb = stadium(P0, P1, rho)
    Th = tensor_l([A1, A2], p, lam, smin=1/rho) - tensor_l([A1], p, lam, smin=1/rho) - tensor_l([A2], p, lam, smin=1/rho)
    Tr = -np.einsum('mij,mj->mi', Th, nn)
    F = {L: (Tr[lb == L]*dl[lb == L, None]).sum(0) for L in ('inner', 'outer', 'vertex', 'tip')}; F['wing'] = (Tr*dl[:, None]).sum(0)
    return F

r0_ = 0.3; A1 = panels(r0_*u1, 3*u1, 220); A2 = panels(r0_*u2, 3*u2, 220)
for lam in (None, 10.0):
    R = {rho: wing_parts_int(A1, A2, r0_*u1, 3*u1, rho, lam) for rho in (0.03, 0.015)}
    print('lambda =', lam)
    for L in ('inner', 'outer', 'vertex', 'tip', 'wing'):
        ex = 2*R[0.015][L] - R[0.03][L]
        print('  %-7s rho=0.03 (%+.4f, %+.4f)  rho=0.015 (%+.4f, %+.4f)  -> rho=0 (%+.3f, %+.3f)' % ((L,) + tuple(R[0.03][L]) + tuple(R[0.015][L]) + tuple(ex)))
rr = np.linspace(0.3, 3, 200001); pp = pi**2/(480*(2*rr*np.sin(a))**4)
print('PFA (Антипин) для внутренней грани, идеальный металл: (%+.3f, %+.3f)' % (-trapz(pp*np.cos(a), rr), trapz(pp*np.sin(a), rr)))
```

|---|---|---|---|
| внутренняя грань (щель) | +0,465 | +0,678 | **+0,89** (PFA: +0,914) |
| внешняя грань | −0,245 | −0,351 | **−0,46** |

**Что это значит.** Давление в щели, которое считает Антипин, реально: тензор воспроизводит PFA для внутренней грани с точностью ~3 %. Но тензор показывает и то, чего в PFA нет. Поле взаимодействия создаёт противоположную по $y$ силу примерно поровну на **внешней грани** (снаружи уголка давление вакуума тоже изменено присутствием второго крыла) и на **торце у вершины**. ## 12c. То же для толстых крыльев ($t=0{,}08$)

У толстого крыла (§10, прямоугольник $t=0{,}08$, зазор у вершины 0,075) это четыре отдельные грани. Контур — тот же, что в §10.2 (`hug`), тензор — взаимодействия, как в §12b. Метки: `side+` — внутренняя грань (в щели), `side-` — внешняя, `vertex` — торец у вершины, `tip` — свободный торец.

## 12d. Профиль давления взаимодействия вдоль граней (толстые крылья)

Давление взаимодействия считаем точка за точкой: $p=-n_i\,\sigma^{\rm int}_{ij}n_j$, где $\sigma^{\rm int}$ — натяжение из тензора взаимодействия §12b, $n$ — внешняя нормаль крыла. Точки лежат на расстоянии $\varepsilon$ от внутренней грани, от внешней грани и поперёк торца у вершины. Знак: $p<0$ — поле тянет грань наружу (притяжение). Для сравнения — PFA Антипина $\pi^2/(480\,d(r)^4)$ при локальном зазоре $d(r)=2r\sin\alpha-t$.

```sage

def face_profile(eps, Ns=60, t=0.08, r0_=0.3, r1_=3.0, lam=None):
    W1 = rect_wing(r0_*u1, r1_*u1, t, 140, 16); W2 = rect_wing(r0_*u2, r1_*u2, t, 140, 16)
    n1 = np.array([-u1[1], u1[0]])          # +n1 смотрит в щель (к крылу 2)
    L = r1_ - r0_
    s = r0_ + L*(1 - np.cos(np.linspace(0, pi, Ns+2)[1:-1]))/2      # гуще к концам
    P0 = r0_*u1
    pin  = np.outer(s, u1) + n1*(t/2 + eps); nin  = np.tile(n1, (Ns, 1))
    pout = np.outer(s, u1) - n1*(t/2 + eps); nout = np.tile(-n1, (Ns, 1))
    q = np.linspace(-t/2, t/2, 9)[1:-1]
    pv = P0 - u1*eps + np.outer(q, n1); nv = np.tile(-u1, (len(q), 1))
    pts = np.vstack([pin, pout, pv]); nrm = np.vstack([nin, nout, nv])
    Th = tensor_l([W1, W2], pts, lam, smin=1/eps) - tensor_l([W1], pts, lam, smin=1/eps) - tensor_l([W2], pts, lam, smin=1/eps)
    tr = -np.einsum('mij,mj->mi', Th, nrm)              # сила на единицу длины, действующая на крыло
    pn = -np.einsum('mi,mi->m', tr, nrm)                # давление: >0 — давит на металл
    return s, pn[:Ns], pn[Ns:2*Ns], q, pn[2*Ns:], tr

import matplotlib.pyplot as plt
PR = {eps: face_profile(eps) for eps in (0.02, 0.01)}
fig, ax = plt.subplots(1, 3, figsize=(15, 4.6))
for eps, ls in ((0.02, '--'), (0.01, '-')):
    s, pin, pout, q, pv, _ = PR[eps]
    ax[0].loglog(s, -pin, ls, color='C0', label='внутр. грань, eps=%.2f' % eps)
    ax[0].loglog(s, np.abs(pout), ls, color='C2', label='|внешн. грань|, eps=%.2f' % eps)
    ax[2].plot(q/0.04, -pv, ls + 'o', color='C3', ms=3, label='eps=%.2f' % eps)
s, pin, pout, q, pv, _ = PR[0.01]; ppfa = pi**2/(480*(2*s*np.sin(a) - 0.08)**4)
ax[0].loglog(s, ppfa, ':k', label='PFA (Антипин)'); ax[0].legend(fontsize=8); ax[0].set_xlabel('r от вершины')
ax[1].semilogx(s, -pin/ppfa, color='C0', label='внутр./PFA'); ax[1].semilogx(s, pout/ppfa, color='C2', label='внешн./PFA')
ax[1].axhline(1, color='k', lw=.5); ax[1].axhline(0, color='k', lw=.5); ax[1].legend()
ax[2].set_xlabel('поперёк торца: -1 внешняя кромка ... +1 кромка у щели'); ax[2].legend()
plt.tight_layout(); plt.savefig('corner_face_pressure_profile.png', dpi=130); plt.show()
```

**Что видно (у меня):**
* **Внутренняя грань** на отрезке $r\approx0{,}4\ldots1$ идёт по PFA с точностью несколько процентов: давление Антипина в щели настоящее. Ближе к вершине ($r<0{,}4$) давление *ниже* PFA (0,57 при $r=0{,}30$): поле вытекает из щели через её узкий конец. При $r>1$ отношение «пилит» 1,0–1,6. Это дискретность: в середине крыла шаг панелей (~0,03) больше $\varepsilon=0{,}01$, и точка видит отдельные панели. Физического смысла у зубцов нет.
* **Внешняя грань** под давлением взаимодействия тоже: $|p_{\rm out}|$ — от ~0,2 % PFA у вершины до ~10 % у раскрыва, по знаку тоже притяжение. В интеграле это мало ($F_y\approx-0{,}2\ldots-0{,}3$).
* **Торец у вершины** — основной носитель встречной силы. Давление на нём большое (до ~130, сравнимо с давлением в щели у вершины ~350) и нарастает к кромке, примыкающей к щели. Торец «видит» второе крыло через узкий конец щели, и поле тянет его к вершине. Рост с уменьшением $\varepsilon$ у этой кромки (60→130) — угловая особенность, где сходятся щель и торец; поэтому в §12c вклады грани и торца перетекают друг в друга.

## 12e. Сплошной уголок (одно тело, как в патенте), ТМ и ТЕ

**Геометрия.** Один замкнутый шестиугольник: два крыла толщины $t=0{,}08$ под $\pm15^\circ$ к оси $y$, длиной до $r_1=3$, сросшиеся у вершины. Внутренний (вогнутый) угол — дно канавки — скруглён радиусом $\rho_{\rm in}$, внешний острый угол вершины — радиусом $\rho_{\rm out}$. Кромки торцов острые. Зазора и вычитаемых одиночных тел нет, поэтому здесь используется **полный** тензор (рассеянная часть $g_s=g-g_0$), а не тензор взаимодействия.

**Чистые ТМ и ТЕ — точные, даже с наклонными крыльями.** Пусть поле $\propto e^{ik_zz}$. Поперечные компоненты выражаются через продольные:
$$
\mathbf E_\perp=\frac{i}{\gamma^2}\big(k_z\nabla_\perp E_z-\omega\,\hat z\times\nabla_\perp H_z\big),\qquad \gamma^2=\omega^2-k_z^2 .
$$
Касательная к сечению составляющая на границе $E_\tau=\frac{i}{\gamma^2}\big(k_z\partial_\tau E_z+\omega\,\partial_nH_z\big)$. Условие идеального проводника $E_z=E_\tau=0$: из $E_z|_\Gamma=0$ следует $\partial_\tau E_z|_\Gamma=0$, и остаётся $\partial_nH_z|_\Gamma=0$. Условия на $E_z$ и $H_z$ **развязаны** при любой форме сечения, в том числе с наклонными гранями и скруглениями. Смешанных мод нет. Смешивание появляется, только если нарушена хотя бы одна из предпосылок: (а) тело конечное вдоль $z$ (торцы по $z$); (б) металл неидеальный — импедансное условие $E_\tau=Z\,H_z$ при $k_z\neq0$ связывает $E_z$ и $H_z$; (в) диэлектрик или анизотропия.

```sage

def rounded_polygon(V, rad, Nline, Narc=16):
    """Замкнутый CCW-многоугольник V с радиусами скругления rad[i] (0 = острый угол).
    Возвращает тело (mid, h, nrm) и описание сегментов для построения контуров."""
    V = [np.asarray(v, float) for v in V]; n = len(V); segs = []; corners = []
    for i in range(n):
        Pp, P, Pn = V[i-1], V[i], V[(i+1) % n]
        din = (P-Pp)/np.linalg.norm(P-Pp); dout = (Pn-P)/np.linalg.norm(Pn-P)
        phi = np.arctan2(din[0]*dout[1]-din[1]*dout[0], din@dout)      # >0 выпуклый
        rho = rad[i]; L = rho*np.tan(abs(phi)/2)
        Pa, Pb = P - din*L, P + dout*L
        left = np.array([-din[1], din[0]])
        C = Pa + np.sign(phi)*rho*left if rho > 0 else P
        corners.append(dict(P=P, Pa=Pa, Pb=Pb, C=C, rho=rho, phi=phi, din=din, dout=dout))
    M = []; H = []; NN = []; LB = []
    for i in range(n):
        c0, c1 = corners[i], corners[(i+1) % n]
        A, B = c0['Pb'], c1['Pa']; m, h = panels(A, B, Nline[i]); tau = (B-A)/np.linalg.norm(B-A)
        M.append(m); H.append(h); NN.append(np.tile([tau[1], -tau[0]], (len(h), 1))); LB += [('L', i)]*len(h)
        segs.append(('line', i, A, B))
        c = c1
        if c['rho'] > 0:
            a0 = np.arctan2(*(c['Pa']-c['C'])[::-1]); sgn = np.sign(c['phi'])
            th = a0 + sgn*np.linspace(0, abs(c['phi']), Narc+1)
            pts = c['C'] + c['rho']*np.stack([np.cos(th), np.sin(th)], 1)
            mm = (pts[1:]+pts[:-1])/2; hh = np.linalg.norm(pts[1:]-pts[:-1], axis=1)
            tt = (pts[1:]-pts[:-1])/hh[:, None]
            M.append(mm); H.append(hh); NN.append(np.stack([tt[:, 1], -tt[:, 0]], 1)); LB += [('A', (i+1) % n)]*Narc
    return (np.vstack(M), np.concatenate(H), np.vstack(NN)), corners, np.array(LB, dtype=object)

def offset_contour(corners, eps, nq=8, narc=24):
    """Контур на расстоянии eps снаружи тела; метки ('L',i) — грань i (V[i]->V[i+1]), ('A',i) — угол i."""
    n = len(corners); P = []; N_ = []; DL = []; LB = []
    for i in range(n):
        c0, c1 = corners[i], corners[(i+1) % n]
        A, B = c0['Pb'], c1['Pa']; Ls = np.linalg.norm(B-A); tau = (B-A)/Ls; nr = np.array([tau[1], -tau[0]])
        s, w = graded(Ls, eps, nq)
        P.extend(A + nr*eps + np.outer(s, tau)); N_.extend([nr]*len(s)); DL.extend(w); LB += [('L', i)]*len(s)
        c = c1; sgn = np.sign(c['phi'])
        Rr = c['rho'] + sgn*eps if c['rho'] > 0 else eps
        nin = np.array([c['din'][1], -c['din'][0]])
        a0 = np.arctan2(nin[1], nin[0]); xg, wg = np.polynomial.legendre.leggauss(narc)
        th = a0 + sgn*(xg+1)/2*abs(c['phi'])
        d = np.stack([np.cos(th), np.sin(th)], 1)
        if c['rho'] > 0 and sgn < 0:            # вогнутый угол: центр скругления снаружи тела
            P.extend(c['C'] - Rr*d)
        else:
            P.extend(c['C'] + Rr*d if c['rho'] > 0 else c['P'] + eps*d)
        N_.extend(d); DL.extend(Rr*wg/2*abs(c['phi'])); LB += [('A', (i+1) % n)]*narc
    return np.array(P), np.array(N_), np.array(DL), LB

def solid_corner(alpha_deg=15, t=0.08, r1=3.0, rho_in=0.02, rho_out=0.02, rho_tip=0.0, Nl=140, Nt=16, Narc=16):
    a = np.radians(alpha_deg); u1 = np.array([np.sin(a), np.cos(a)]); u2 = np.array([-np.sin(a), np.cos(a)])
    g1 = np.array([-np.cos(a), np.sin(a)]); g2 = np.array([np.cos(a), np.sin(a)])     # в щель от крыла 1 / крыла 2
    rI = t/(2*np.tan(a)); I = rI*u1 + g1*t/2; O = -rI*u1 - g1*t/2
    V = [O, r1*u1 - g1*t/2, r1*u1 + g1*t/2, I, r1*u2 + g2*t/2, r1*u2 - g2*t/2]
    names = ['внешняя грань 1', 'торец 1', 'внутренняя грань 1', 'внутренняя грань 2', 'торец 2', 'внешняя грань 2']
    cnames = ['внешний угол (вершина)', 'кромка торца 1 внеш.', 'кромка торца 1 внутр.', 'внутренний угол (дно)', 'кромка торца 2 внутр.', 'кромка торца 2 внеш.']
    rad = [rho_out, rho_tip, rho_tip, rho_in, rho_tip, rho_tip]
    body, corners, lb = rounded_polygon(V, rad, [Nl, Nt, Nl, Nl, Nt, Nl], Narc)
    return body, corners, names, cnames

def chunked(fun, bodies, pts, chunk=120, **kw):
    return np.vstack([fun(bodies, pts[i:i+chunk], **kw) for i in range(0, len(pts), chunk)])

def circle_flux(body, center, Rc, pol, M=400, nk=60):
    th = 2*pi*np.arange(M)/M; n = np.stack([np.cos(th), np.sin(th)], 1); pts = center + Rc*n
    if pol == 'TM': Th = tensor_l([body[:2]], pts, None, nk=nk)
    else:           Th = chunked(tensorN, [body], pts, nk=nk, smin=None)
    return -(np.einsum('mij,mj->mi', Th, n)*(2*pi*Rc/M)).sum(0)

def parts_flux(body, corners, eps, pol, nk=60):
    p, nn, dl, lb = offset_contour(corners, eps)
    if pol == 'TM': Th = tensor_l([body[:2]], p, None, nk=nk, smin=1/eps)
    else:           Th = chunked(tensorN, [body], p, nk=nk, smin=1/eps)
    Tr = -np.einsum('mij,mj->mi', Th, nn)
    out = {}
    for k in set(lb):
        msk = np.array([l == k for l in lb]); out[k] = (Tr[msk]*dl[msk, None]).sum(0)
    return out, (Tr*dl[:, None]).sum(0)

def show(res, tot, lab, names, cnames):
    print(lab)
    for k in sorted(res, key=lambda x: (x[0], x[1])):
        nm = names[k[1]] if k[0] == 'L' else cnames[k[1]]
        print('   %-26s (%+11.4f, %+11.4f)' % (nm, *res[k]))
    print('   %-26s (%+11.4f, %+11.4f)' % ('ИТОГО', *tot))

cen = np.array([0.0, 1.5])
# 1) поток по окружностям, ТМ и ТЕ, разные скругления
for (ri, ro) in ((0.02, 0.02), (0.005, 0.005), (0.02, 0.0)):
    b, c_, nm, cn = solid_corner(rho_in=ri, rho_out=ro)
    for pol in ('TM', 'TE'):
        for Rc in (2.2, 3.0):
            F = circle_flux(b, cen, Rc, pol, nk=50)
            print('rho_in=%.3f rho_out=%.3f  %s  R=%.1f:  F = (%+.2e, %+.2e)' % (ri, ro, pol, Rc, *F))
# 2) раскладка по граням и скруглениям: панели мельче eps, rho_in > eps
b, c_, nm, cn = solid_corner(rho_in=0.06, rho_out=0.02, Nl=200, Nt=20, Narc=20)
res, tot = parts_flux(b, c_, 0.04, 'TM', nk=40); show(res, tot, 'TM, eps=0.04, rho_in=0.06, rho_out=0.02', nm, cn)
```

**Результат (у меня).**

Поток по окружности (вариант 1), $\rho_{\rm in}=\rho_{\rm out}=0{,}02$:

| поляризация | $R=2{,}2$ | $R=3$ |
|---|---|---|
| ТМ | $F_y=1{,}9\cdot10^{-10}$ | $9{,}7\cdot10^{-12}$ |
| ТЕ | $F_y=6{,}7\cdot10^{-6}$ | $1{,}5\cdot10^{-6}$ |

ТМ при $\rho_{\rm in}=\rho_{\rm out}=0{,}005$ и при $\rho_{\rm out}=0$: $F_y\approx1{,}3$–$1{,}5\cdot10^{-10}$. Остаток падает с ростом $R$ и от радиусов скругления не зависит. Это численный ноль. Для ТЕ он грубее, как и в §11.

Раскладка по поверхности (вариант 2, ТМ, $\varepsilon=0{,}04$, $\rho_{\rm in}=0{,}06$): вклады отдельных частей до ~56 — это самодействие острых кромок и скруглений одного тела. Итог $F_y=+0{,}037$, $F_x=0$, то есть ноль с точностью ~$10^{-3}$ от частей.

**Две предосторожности, без которых получаются бессмысленные числа** (сначала я на них попался):
1. Контур у вогнутого скругления проходит на радиусе $\rho_{\rm in}-\varepsilon$, поэтому нужно $\varepsilon<\rho_{\rm in}$. Иначе контур выворачивается внутрь металла: при $\rho_{\rm in}=0{,}02$, $\varepsilon=0{,}04$ «сила на дно» получалась $+2{,}4\cdot10^5$.
2. Шаг панелей должен быть меньше $\varepsilon$. При $h_{\max}=0{,}03$ и $\varepsilon=0{,}01$ части доходили до $10^5$, а сумма — до $4\cdot10^4$.

**Что здесь не определено.** У одного тела нет тензора взаимодействия, поэтому «давление Антипина в канавке» отдельно от самодействия формы не выделяется: оба вклада одного порядка у дна и у кромок. Физически устойчивы только сумма по замкнутому контуру и её независимость от $\rho_{\rm in},\rho_{\rm out}$.

### 12e.2. ТЕ: сходимость суммы по поверхности при сгущении сетки

Вопрос: остаток $F_y$ на прижатом контуре — сила (скачок тензора на металле) или погрешность? Контур на расстоянии $\varepsilon$ охватывает скачок целиком при любом $\varepsilon>0$, а между двумя такими контурами $\partial_j\Theta_{ij}=0$. Поэтому физическая сила не может зависеть ни от $\varepsilon$, ни от сетки. Проверка: фиксируем $\varepsilon$ и сгущаем панели.

Внимание: расчёт с $N_l=600$ требует много памяти (у меня — на Mac, ~3 мин на точку).

```sage
for eps in (0.04, 0.03):
    for Nl in (200, 300, 450, 600):
        b, c_, nm, cn = solid_corner(rho_in=0.06, rho_out=0.02, Nl=Nl, Nt=max(20, Nl//10), Narc=max(20, Nl//12))
        res, tot = parts_flux(b, c_, eps, 'TE', nk=40)
        print('eps=%.2f  Nl=%d  h_max=%.4f  ИТОГО F_y = %+.4f' % (eps, Nl, b[1].max(), tot[1]))
```

|---|---|---|---|
| 200 | 0,024 | +1,21 | — |
| 300 | 0,016 | +0,75 | +9,4 |
| 450 | 0,0107 | +0,42 | +1,21 |

(Строки $N_l=200,300$ считались с $N_t=20/30$, $N_{\rm arc}=20/30$; в ячейке выше параметры для них немного другие, поэтому числа могут чуть отличаться.)

Остаток монотонно падает при сгущении, в среднем как $h^{1{,}3}$. Подгонка $F_y=F_\infty+Ch^p$ по трём последним точкам $\varepsilon=0{,}04$ даёт $p\approx1{,}7$, $F_\infty\approx+0{,}09$; по двум точкам $\varepsilon=0{,}03$ с тем же $p$ — $F_\infty\approx+0{,}27$. Разные $\varepsilon$ дают разные пределы. Физическая сила так себя вести не может, значит и эти $\sim0{,}1$ — остаток экстраполяции. Масштаб частей ~60–160, то есть остаток $\lesssim10^{-3}$ от частей. Поток по окружности для того же тела — $10^{-6}$.

**Вывод:** в ТЕ однотельная сила на сплошной уголок не обнаружена. Остаток на прижатом контуре сходится к нулю при сгущении сетки, со скоростью $h^{1{,}3}$–$h^{1{,}7}$; экстраполированный предел ($0{,}1$–$0{,}3$) зависит от $\varepsilon$ и поэтому сам является погрешностью экстраполяции. Строгий ноль численно не достигнут: ТЕ-решатель с кусочно-постоянной плотностью сходится медленно.

## 13. Итоги

1. **Метод корректен.** $K_0$ и самопанели сверены с прямыми интегралами. Условие Дирихле выполняется. $\partial_j\Theta_{ij}=0$ проверено численно. Сила через тензор совпадает с энергетическим методом на двух телах (тест 9) и на каждом крыле уголка (10.3, расхождение доли процента; оно уменьшается с ростом числа панелей).
2. **Вариант 1 (окружность).** Полная сила на уголок порядка $10^{-11}$–$10^{-14}$ от силы сжатия крыльев, то есть ноль на уровне численной точности, при всех $R$.
3. **Вариант 2 (поверхность).** Сила на каждое крыло $\approx(\mp35{,}2;\ 0)$ — это сжатие крыльев друг к другу. Она не зависит от $\varepsilon$ и совпадает с энергетическим методом. Вклады граней и торцов по отдельности зависят от $\varepsilon$ и от сетки (10.4), поэтому физического смысла не имеют. Сумма по двум крыльям — ноль: $|F|/|F_x^{(1)}|\sim3\cdot10^{-7}$ при $\varepsilon=0{,}04$. Ошибка растёт при уменьшении $\varepsilon$ ($2\cdot10^{-3}$ при $\varepsilon=0{,}01$), потому что контур подходит к панелям ближе их размера. Это численный эффект, не тяга.
4. **Почему варианты обязаны совпадать.** Между поверхностью крыльев и окружностью $\partial_j\sigma_{ij}=0$ (§6), поэтому по Гауссу (§7) это один и тот же интеграл. Поверхностный контур не может дать другую полную силу. Он даёт лишь раскладку: грани давят наружу, торцы у вершины — внутрь. Это та же картина, что в ч. III `Antipin_angle_finite_DC.ipynb`.
5. **ТЕ (Нейман, §11)** — тот же итог: ноль полной силы по окружности, сжатие крыльев.
6. **Полупрозрачный металл и почти сомкнутая вершина (§12)** — сила сжатия меняется на порядки, суммарная остаётся нулём.
7. **Сплошной уголок, одно тело (§12e)** — со скруглением внутреннего и внешнего углов, ТМ и ТЕ: поток по окружности — численный ноль, не зависит от радиусов скругления. Чистые ТМ/ТЕ точны при любой форме сечения идеального проводника.
8. **Чего здесь нет:** дисперсии $\varepsilon(\omega)$, $T>0$, неравновесия и движения. Для них равенство нулю отдельно не проверено.

**Как самому проверить точность:** увеличить число панелей в `rect_wing` (140→200, 16→24) и `nk` (60→80). Результаты 10.0–10.3 должны меняться в соответствующих знаках.
